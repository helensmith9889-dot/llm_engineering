````python
# ========== 环境：从 .env 加载 Anthropic API Key 并做基本校验 ==========

# override=True：.env 里的值覆盖进程里已有的同名环境变量
load_dotenv(override=True)
# 从环境变量读取 Anthropic 密钥；名字必须是 ANTHROPIC_API_KEY（与 .env 键名一致）
api_key = os.getenv('ANTHROPIC_API_KEY')

# ---------- 三项体检：有没有 Key / 前缀对不对 / 有没有首尾空白 ----------

# 情况 1：完全没读到 Key
if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
# 情况 2：有值但不是 Anthropic 常见前缀 sk-ant-
elif not api_key.startswith("sk-ant-"):
    print("An API key was found, but it doesn't start sk-ant-; please check you're using the right key - see troubleshooting notebook")
# 情况 3：首尾可能有空格/制表符（复制粘贴常见坑）
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
# 情况 4：目前看起来正常（真正能否调用还要看下一格 API 请求）
else:
    print("API key found and looks good so far!")
````

# 预览：先快速调用一次 Frontier 模型，感受 Messages API


````python
# ========== 预览：组装 messages 列表（还没真正发请求） ==========

# 用户说的一句话：发给模型的 content 保持英文（影响回答的字符串不翻译）
message = "Hello, Claude! This is my first ever message to you! Hi!"

# Anthropic / 多数 Chat API 都用「消息列表」：每条有 role（角色）和 content（内容）
# 这里只有一条 user 消息；系统提示（system）后面单独用参数传
messages = [{"role": "user", "content": message}]

# 在笔记本里单独写变量名：会显示该对象，方便你检查结构
messages
````

````python
# ========== 真正调用 Claude：创建客户端 → messages.create → 取回复文本 ==========

# 创建 Anthropic 客户端；默认会从环境变量 ANTHROPIC_API_KEY 读密钥
client = Anthropic()

# messages.create：同步调用；model / max_tokens / messages 等参数保持原样（勿改 model id）
response = client.messages.create(model="claude-opus-4-8", max_tokens=1024, messages=messages)
# Claude 返回的 content 是块列表；取第一块的 .text 即助手回复字符串
response.content[0].text
````

## 提示类型（Prompt Types）

你可能已经知道——若还不知道，很快就会非常熟悉！

像 GPT / Claude 这类模型，训练时就约定了「怎么接收指令」。它们通常期望两类输入：

- **系统提示（System Prompt）**：告诉模型在执行什么任务、该用什么语气/风格  
- **用户提示（User Prompt）**：对话的起点——用户真正想问/想处理的内容


````python
# ========== 定义 System Prompt：定「人设 + 任务 + 输出格式」 ==========

# 可稍后实验：把最后一句改成要求用西班牙语 Markdown 回复等
# 注意：下面三引号里的英文是发给模型的指令，翻译会改变回答风格，故保留英文
system_prompt = """
You are a snarky assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
````

````python
# ========== 定义 User Prompt 前缀：后面会拼上网页正文 ==========

# 这是「用户侧」说明：告诉模型正文来了、要做什么；具体网站内容在后面拼接
user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""
````

## 消息（Messages）在 Anthropic API 里怎么传

在 **Anthropic** 的 API 中：系统提示是单独的 `system=` 字符串参数——这里**不需要** `role` 键，因为只有一个 system，用途是隐含的：

```python
system = "You are a helpful assistant"  # just a string, no role key
```

但消息列表里的 `role` 键仍然必需——只是不再包含 `"system"` 选项，通常只有 `"user"` 与 `"assistant"`：

```python
messages = [
    {"role": "user", "content": "What is 2 + 2?"}  # role key still required here
]
```


````python
# ========== 小例子：system 定海盗口吻，user 问算术 ==========

# messages：对话内容；这里只有一条 user
messages = [
        {"role": "user", "content": "What is 2 + 2?"}
]

# system= 单独传人设；model / 文案保持原样（影响行为的字符串不改）
response = client.messages.create(model="claude-opus-4-8",max_tokens=1024,system="You are a pirate. Respond only in pirate speak.",
    messages=messages
)
# 取出助手回复文本
response.content[0].text
````

## 用函数为后续摘要流程组装 messages

下面把「用户提示前缀 + 网页正文」封装成函数，方便反复调用（模型名在后面的 `summarize` 里指定）。


````python
# ========== 辅助函数：把网页文本变成 API 需要的 messages 格式 ==========

# 返回值结构与上面示例一致：[{role:user, content:前缀+正文}]
def messages_for(website):
    return [
        # content = 固定前缀说明 + 抓取到的网站文本
        {"role": "user", "content": user_prompt_prefix + website}
    ]
````

````python
# ========== 试跑：看看 messages_for 拼出来长什么样 ==========

# 用前面抓到的 ed（edwarddonner.com 正文）做输入；也可换成别的网站文本再试
messages_for(ed)
````

## 拼起来：抓取 → 组消息 → 调 Anthropic API

Anthropic 的 Messages API 调用模式很直接，后面课程会反复用到。


````python
# ========== 核心函数 summarize：URL → 抓取 → Claude 摘要 ==========

# 端到端流水线：你以后会非常熟悉这种「工具函数 + LLM」组合
def summarize(url):
    # 第一步：用 scraper 把网页变成纯文本
    website = fetch_website_contents(url)
    # 第二步：调用 Claude；system 定口吻，messages 带网页内容
    response = client.messages.create(
        model = "claude-sonnet-4-6",
        max_tokens=1024,
        system="You are a glamour girl. Respond only in glamour girl speak.",
        messages = messages_for(website)
    )
    # 第三步：从响应块里取出文本返回
    return response.content[0].text
````

````python
# ========== 调用 summarize：对课程作者个人站做一次摘要 ==========

summarize("https://edwarddonner.com")
````

````python
# ========== 展示函数：把摘要渲染成 Markdown，笔记本里更好读 ==========

def display_summary(url):
    # 先拿到模型生成的 Markdown 字符串
    summary = summarize(url)
    # display(Markdown(...))：在 Jupyter 里按 Markdown 渲染，而不是纯文本
    display(Markdown(summary))
````

````python
# ========== 带渲染地再跑一次同一站点 ==========

display_summary("https://edwarddonner.com")
````
