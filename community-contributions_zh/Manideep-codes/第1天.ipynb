```python
# ========== 第 1 天练习：用 Chat Completions 给邮件建议主题 ==========

# 导入标准库 os：读环境变量（Environment Variables），例如 OPENAI_API_KEY
import os
# 从 dotenv 导入 load_dotenv：把 .env 里的密钥读进进程环境，避免把密钥写进代码
from dotenv import load_dotenv
# 从 openai 导入 OpenAI 客户端类（本格后面创建了 client，但实际调用用的是 openai 模块级 API）
from openai import OpenAI
# 导入 openai 模块：后面用 openai.chat.completions.create(...)
import openai
# 创建 OpenAI 客户端实例（默认会读 OPENAI_API_KEY；本练习主路径仍走下面的 openai.chat...）
client = OpenAI()

# 加载 .env；override=True 表示用文件里的值覆盖已有环境变量
load_dotenv(override=True)
# 取出 API Key（本格读取后未直接传入 create；依赖环境/客户端默认行为）
api_key = os.getenv('OPENAI_API_KEY')

# ========== 步骤 1：创建提示词（发给模型的英文字符串保持原样）==========

# system：设定助手角色——帮用户给邮件想主题（subject）
system_prompt = "You are a helpful assistant to suggest subject of the emails"
# user：一封求职意向邮件正文；模型要据此建议邮件主题
user_prompt = """
    Dear Jasmine,

I hope you're doing well. I came across your job postings for the Full Stack (Java/React) Developer and Backend (Java) Developer roles, and I am very interested in exploring these opportunities.

With my experience in Java development, particularly with Spring Boot, Microservices, API development, and cloud platforms, I believe I would be a strong fit for either role. I have previously worked on enterprise-level applications, handling Kafka for messaging, and deploying solutions on AWS and OpenShift.

I would love to discuss how my skills align with your needs. Please let me know a convenient time for a quick chat. I’ve attached my resume for your reference.

Looking forward to your response!


"""

# ========== 步骤 2：构造 messages 列表（Chat Completions 标准结构）==========

# messages：system 定角色，user 放具体邮件内容
messages = [{"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}] # fill this in

# ========== 步骤 3：调用 OpenAI Chat Completions API ==========
# model 名字保持原样；messages 即上一步组装的对话
response = openai.chat.completions.create(
    model = "gpt-4.1-mini",
    messages = messages
)
# response =

# ========== 步骤 4：打印模型回复（主题建议在 choices[0].message.content）==========
print(response.choices[0].message.content)
# print(
```
