# 第 3 周练习 —— PDF 诗集情绪分析器

## 练习目标（理念）

对一本**免费诗集 PDF**做文本情绪（Sentiment）分析：按诗拆分，估计每首的情感倾向，再用**条形图**与**折线图**可视化结果。

本笔记本本身是**入口说明**：完整可运行代码托管在下方 Colab 链接；在此阅读目标与资源即可跳转动手。

## 和本课第 3 周概念的关系

| 本课概念 | 本练习里通常会用到 |
|----------|------------------|
| 文档 / PDF 读取 | 从诗集 PDF 抽取文本 |
| 文本分段 | 按「一首诗」切分后再分析 |
| 情绪分析 | 分类或打分（正面 / 负面 / 中性等） |
| 可视化 | 条形图、折线图对比各诗情绪 |
