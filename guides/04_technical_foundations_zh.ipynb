## 主题 1：ChatGPT 与 OpenAI API

### ChatGPT 和 GPT API 有什么区别？二者都由 OpenAI 提供。

#### ChatGPT 是面向终端用户的工具。它是为使用 AI 的消费者设计的聊天产品。
- 它有免费计划，也有功能更多的付费订阅计划。
- 订阅计划让用户近乎不受限地使用该聊天产品。

#### API 是为 AI 工程师——从事其他商业产品的软件工程师和数据科学家——提供的服务。
- 它让像你我这样的技术人员可以访问底层模型（如 “GPT4.1” 和 “o3”），从而构建自己的产品。
- 如果我们愿意，可以用 API 构建自己的 ChatGPT 版本，并向终端用户收费。
- 与大多数 API 一样，OpenAI 按 API 用量收取少量费用。课程中多数使用 gpt-4o-mini 的示例，单次 API 调用大约是 $0.001 量级。

### 我已经每月为 ChatGPT 付 $20——为什么还要为 API 额外付费？

- 希望现在已经清楚了。API 不是给消费者的；它是给工程师用来构建可收费平台的。
- 如果你能凭订阅获得 API 访问权，就可以用更低价格向他人提供 ChatGPT 类工具，那会把 OpenAI 挤垮！
- 请记住：每次 API 调用可能需要 10,000,000,000,000 次浮点运算——这些计算要耗电！

你也可以不调用 API，而是在本地运行开源模型，但通常它们的计算量大约少 1,000 倍——即便很小，这些处理仍然会计入你的电费。
