# Guerilla Search — 本地法学硕士的网站分析

Guerilla Search 是一种网站分析工具，它使用通过 Ollama 本地运行的大型语言模型来超越传统的网页摘要。

给定网站 URL，系统提取网页内容并生成结构化分析，不仅探索网站传达的内容，还探索周围的上下文和该通信背后可能的观点。

分析内容包括：

* **官方故事** — 网站明确呈现的内容
* **隐藏背景** — 可能的假设、激励和框架
* **缺失的声音** - 可能无法代表的观点
* **公众反应** - 来自不同受众的假设观点
* **游击队判决** - 简洁而诙谐的最终观察

该项目旨在通过将可观察信息与解释分开并避免未经支持的结论，鼓励对在线内容进行更深思熟虑的阅读。

该实现完全在使用 Ollama 的本地法学硕士上运行，同时保留与 OpenAI SDK 接口的兼容性，无需使用外部 API 即可进行实验。

```python
from openai import OpenAI
import sys
sys.path.append("../..")
from week1.scraper import fetch_website_contents
import requests

requests.get("http://localhost:11434").content
```

```python
OLLAMA_BASE_URL = "http://localhost:11434/v1"

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')
```

```python
system_prompt = """
You are Guerilla Search.

Your goal is NOT to summarize websites.

Your goal is to reveal hidden context and societal undercurrents.

For every website:

1. Explain the official message.
2. Infer possible motivations.
3. Identify missing perspectives.
4. Mention who benefits.
5. Mention groups that may disagree.
6. Point out assumptions or marketing language.
7. Add mild humor and irony.

Never invent facts.
Separate:
- Facts
- Inferences
- Speculation

Respond in markdown.
"""
```

```python
user_prompt_prefix = """
Analyze this website.

Website Content:
"""
```

```python
website = fetch_website_contents("https://wikipedia.org")

context = f"""
{website}

Instructions:
- Official Story → facts only
- Hidden Context → cautious inference
- Missing Voices → only if noticeable
- Public Reaction → hypothetical
- Never invent events
"""

response = ollama.chat.completions.create(
    model="llama3.2",
    messages=[
        {
            "role":"system",
            "content":system_prompt
        },
        {
            "role":"user",
            "content":user_prompt_prefix + context
        }
    ]
)

print(response.choices[0].message.content)
```
