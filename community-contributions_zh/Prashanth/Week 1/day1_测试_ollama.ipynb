```python
# ========== 用 OpenAI 兼容接口调用本地 Ollama ==========
# 思路：Ollama 提供 /v1（OpenAI-compatible），所以 SDK 用法和云端几乎一样，只换 base_url

# 从 openai 导入 OpenAI 客户端类：后面用 chat.completions.create 发对话
from openai import OpenAI
# 本地模型名：必须和 ollama list / 已 pull 的名字一致
MODEL = "llama3.2"
# 从 IPython.display 导入 Markdown、display：在笔记本里渲染 Markdown 回答
from IPython.display import Markdown, display
# 创建客户端：指向本机 Ollama 的 OpenAI 兼容端点；api_key 对本地常填占位 "ollama"
openai = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")

# 发起一次非流式 Chat Completions：等模型整段生成完再返回
response = openai.chat.completions.create(
 model=MODEL,
 # messages：system 定回答风格，user 放具体问题（prompt 字符串保持英文，改译会改模型行为）
 messages=[{"role": "system", "content": "Respond concisely, use bullet points"},{"role": "user", "content": "give information about a proper diet"}]
)

# 调试草稿（中文音译注释残留，保持不执行）
# 打印（响应.选择[0].消息.内容）
# 打印（响应）

# 取出助手回复文本，用 Markdown 漂亮展示（比 print 更适合笔记阅读）
display(Markdown(response.choices[0].message.content))
# 打印（响应.选择[0].消息.内容）
```
