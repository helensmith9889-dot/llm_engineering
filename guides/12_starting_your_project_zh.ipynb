## 如何开始推进你的想法

所以你有了一个想构建的东西的想法。

恭喜！最重要的部分已经完成了。想法就是一切！

但现在你不确定从哪里开始。我经常被问到这个问题。

我想给你三条关键建议。



## 建议 1：做一名科学家

作为 AI Engineer，你需要戴两顶帽子：软件工程师，以及数据科学家。

在项目初期，至关重要的是摘下软件工程帽子，牢牢戴上数据科学帽子。

很多来自软件工程背景的人（包括我自己！）在这一点上会挣扎。人们很容易滑进舒适区，去问架构、设计、数据管道、可扩展性、部署之类的问题。

这些都是很好的问题，但在早期阶段它们不是关键问题。它们不是真正难的问题。它们不会决定你的想法成败。

对于关键问题，你需要做一名科学家。

问问自己：  
1. 你将如何评估模型的成功？你要衡量什么？  
2. 你有什么数据，还需要什么数据？  
3. 你如何构建小型原型来测试不同技术并理解模型表现？

先处理科学问题——你能用 LLM 实现什么，以及如何实现。

我还建议先在 Notebook 中开始，再进入 Python 模块，以迫使自己以这种科学心态运作。

这对 Agent 项目尤其关键。画一张大的 Agent 架构图、展示智能体如何协作，像技术图一样，这很诱人。但这是软件工程师的思维方式。我建议换个思路：尝试不同方法，研究什么有效，检验假设并迭代。

## 建议 2：拥抱研发（R&D）

好吧，这和建议 1 很相似，但值得再说一遍！

我经常被问到类似的问题：「Ed，我想构建如下方案。我该用模型 A、B 还是 C？该用 Agents、RAG 还是微调？」

我的回答几乎总是：  
- 你应该全部都试！关键是实验；(a) 提出评估标准，(b) 准备一份精选数据集，(c) 测试不同想法并看表现
- 我常常对哪个最好有直觉，但我的直觉经常是错的！别信我：自己试  
- 实验没有替代品。

## 建议 3：梦想要大，但起步要小！

我也经常收到学生发来 200–300 行代码，说「Ed，这表现不好，请修一下」。😂

正如我在调试指南里说的——事情不是这样的！

开始项目时，从小而简单开始至关重要。打磨每一条提示；在每一步上细致工作。确认每一次 LLM 调用都按你期望表现，并迭代输入，直到响应稳定可靠。

如果你始终增量地工作，用小的、可测试的构建块，你就应该对发生的一切有完全清晰的认识。

当然，我喜欢帮人做项目，帮忙修问题也很棒。但如果你增量地推进项目，你应该总是清楚到底哪里表现不好——那样我帮你就会非常高效。

# 总结

你大概已经注意到这三条建议的共同主题。对非数据科学背景的人来说，用这种看似临时的方式处理问题可能很别扭；感觉「很 hacky」、不踏实。但在我看来，这是成为成功 AI 工程师需要掌握的最重要技能：对不确定性感到自在，享受实验，并拥抱做一名科学家。
