# 家庭作业：用本地 Ollama（llama3.2）总结作品集网站

## 练习目标

把第 1 天「网页摘要」从 **OpenAI 云端**换成 **Ollama 本地开源模型**：同一套 OpenAI 兼容 SDK，只改 `base_url` 与 `model`。

不想付费 API 时，后续项目也能复用这套接法。

## 利弊（教学对照）

**好处**
1. 无 API 按量收费——模型开源、跑在本机
2. 页面正文不离开你的电脑（隐私更可控）

**代价**
1. 能力通常弱于 Frontier 云端模型；本机算力/显存会影响速度与质量

## 和本课概念的关系

| 概念 | 本练习 |
|------|--------|
| OpenAI 兼容客户端 | `OpenAI(base_url=..., api_key='ollama')` |
| 本地推理服务 | `http://localhost:11434/v1` |
| 模型名 | `llama3.2`（也可换成 `llama3.2:1b`） |
| 抓取 + messages | `fetch_website_contents` + system/user |

## Ollama 安装回顾

1. 打开 [ollama.com](https://ollama.com) 安装
2. 浏览器访问 [http://localhost:11434/](http://localhost:11434/) —— 应看到类似 “Ollama is running”
3. 若未运行：终端执行 `ollama serve`（有的环境写成 `ollama serve`，勿与视频口误混淆）
4. 另开终端：`ollama pull llama3.2`
5. 若太慢：`ollama pull llama3.2:1b`，并把代码里的 `model="llama3.2"` 改成 `model="llama3.2:1b"`（本注释只说明，不替你改可执行代码）

## 怎么跑

确保 Ollama 已启动且已 pull 模型 → 运行下方代码格 → 阅读返回的作品集摘要。


```python
# ========== 本地 Ollama：OpenAI 兼容接口总结作品集站 ==========

# 从 scraper 导入抓取函数：请求作品集网站并抽出正文
from scraper import fetch_website_contents
# 导入 OpenAI 客户端：这里不连云端，而是指向本机 Ollama 的 /v1 兼容层
from openai import OpenAI

# Ollama 的 OpenAI 兼容基址（必须含 /v1）；端口 11434 是默认
OLLAMA_BASE_URL = "http://localhost:11434/v1"
# api_key 对本地 Ollama 通常任意非空即可；这里按惯例写 'ollama'
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')

# 抓取目标作品集站点正文；URL 保持原文
maddie = fetch_website_contents("https://madelenecampos.com")

# messages：system 定助手语气，user 放入「请摘要」+ 网页正文
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Can you summarise the following website for me? " + maddie}
]

# 调用本地 llama3.2；model 字符串必须与 ollama list 里的名字一致
response = ollama.chat.completions.create(model="llama3.2", messages=messages)
# Jupyter 会展示最后一个表达式的值：这里是模型回复文本
response.choices[0].message.content
```
