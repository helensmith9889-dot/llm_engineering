## 有了这些基础——理解 Python 中的 NameError

在 Python 中碰到 NameError 相当常见。有了基础知识，你应该总能有把握地调试 NameError 并找到根因。

如果你在 Notebook 中工作，最常见的原因是你没有按顺序执行所有单元格，所以 Kernel 里还没有某个定义。
