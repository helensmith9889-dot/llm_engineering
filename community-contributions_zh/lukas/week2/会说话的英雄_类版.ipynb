## 练习目标（理念）

把第 1 周里「有名有姓」的英雄，变成真正的 **AI Agent**：把他们放在洞穴入口前，目标明确——决定下一步怎么做。

这对应 **Week 2** 的多 Agent、工具调用（tool use）与编排（orchestration）思路：角色不只是聊天，还能查物品栏、施法、互相交接工具，并在规则不允许时被拒绝。

## 和本课 Week 2 的关系

| 本课概念 | 本项目里你会看到 |
|----------|------------------|
| Agent / 多角色 | 多个 RPG Agent 各自持有聊天历史，互相听与说 |
| Tool use | 列法术书、施法、交接法术书/物品、列物品栏等 |
| 规则/护栏 | 不合 DnD 规则或对象定义时，由另一 Agent 做 allow/deny |
| 多轮编排 | 设定游戏回合数；到上限或出现 stop 信号则结束 |
| 成本可见 | LiteLLM callback 统计 $ 与 token |
| GUI | Gradio 界面；类模型封装；流式（streaming） |

## 应用能做什么

- **生成 RPG Agent**：从文本来源读完整规格；由另一 Agent 做角色摘要（较短描述，并选定名字、职业、物品栏）；按指定数量（此处为 3）逐个创建。
- **使用工具（tools）**：列出法术书中的法术、施法、交接法术书（或其它工具）、列出物品栏并告知他人内容。
- **限制工具用法**：请求动作不符合 DnD 规则或对象定义时会被拒绝（allow/deny 由另一 Agent 提供）。
- **收发消息并各自保留聊天历史**：talk / listen，使 Agent 彼此反应，对话看起来有意义。
- **设定初始游戏回合数**：到达回合上限，或 Agent 解决局面并出现 stop 信号时停止。
- **计算成本**：借助 LiteLLM callback 统计聊天花费的 $ 与 token。
- **切换语言**：英语 / 捷克语（地下城主 Dungeon Master 尚未实现——DM 消息仍保持英文）。
