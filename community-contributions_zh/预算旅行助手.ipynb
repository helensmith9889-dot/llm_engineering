```python
# ========== 导入：把后面要用的工具箱搬进来 ==========

# 导入标准库 os：读环境变量（Environment Variables），例如 OPENAI_API_KEY
import os
# 从 dotenv 导入 load_dotenv：把 .env 里的密钥读进环境变量，避免把密钥写进代码
from dotenv import load_dotenv
# 从 IPython.display 导入展示工具：在笔记本里把模型回复渲染成 Markdown
from IPython.display import Markdown, display
# 从 openai 导入 OpenAI 客户端类：调用云端 Chat Completions API
from openai import OpenAI
```

```python
# ========== 环境变量：加载并自检 API Key ==========

# 加载 .env（override=True：已存在的同名环境变量也会被 .env 覆盖）
load_dotenv(override=True)
# 从环境变量读取 OpenAI 密钥；后面只做检查，真正调用时 SDK 也会读同名变量
api_key = os.getenv('OPENAI_API_KEY')

# 分层检查密钥：缺失 / 前缀不对 / 首尾空白 —— 三种常见配置错误
# Check the key

# 情况 1：根本没读到密钥
if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
# 情况 2：有值但不是 sk-proj- 开头（可能拷错了别的密钥）
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
# 情况 3：首尾有空格/制表符，容易导致鉴权失败
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
# 通过以上检查：至少「看起来」可用（仍需真实 API 调用才能最终确认）
else:
    print("API key found and looks good so far!")
```

```python
# ========== 步骤 1：编写 system / user 提示词（Prompt） ==========
# Step 1: Create your prompts

# system_prompt：角色 + 输出约束（专家、Top 5 免费景点、日程建议、纯 Markdown）
# 注意：下面整段英文字符串是可执行 prompt，不要翻译改写
system_prompt = "You are an expert on budget travelling. You always answer with top 5 free tourist attractions and suggest best days and schedules to visit. Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown."
# user_prompt：具体用户问题；改城市名即可复用同一套 system 设定
user_prompt = "Tell me about Paris" 
```

```python
# ========== 步骤 2：组装 messages 列表（Chat Completions 标准格式） ==========
# Step 2: Make the messages list

# messages：有序对话；先 system 定规则，再 user 放问题
messages = [
    # role=system：告诉模型「你是谁、怎么答」
    {"role": "system", "content": system_prompt},
    # role=user：当前这轮用户输入
    {"role": "user", "content": user_prompt}
]
```

```python
# ========== 步骤 3：创建客户端并调用 OpenAI Chat Completions ==========
# Step 3: Call OpenAI

# 创建 OpenAI 客户端；默认从环境变量读取 OPENAI_API_KEY
openai = OpenAI()

# 发起一次非流式补全：model 指定模型 id，messages 传入上一步组装的对话
response = openai.chat.completions.create(model="gpt-5-nano", messages=messages)
```

```python
# ========== 步骤 4：取出回复正文并以 Markdown 展示 ==========
# Step 4: print the result

# choices[0].message.content：第一条候选回复的文本；用 Markdown() 包一层再 display
display(Markdown(response.choices[0].message.content))
```
