```python
# ========== 导入：把后面要用的工具箱搬进来 ==========
# 练习目标：用 Chat Completions 做「职业方向导师」——给兴趣领域，输出资深工程师建议
# 和本课关系：messages（system/user）、OpenAI 客户端、.env 读密钥
# 怎么跑：先配好 OPENAI_API_KEY，再从上到下 Shift+Enter

# 导入标准库 os：读环境变量（Environment Variables），例如 API Key
import os
# 从 dotenv 导入 load_dotenv：把 .env 里的密钥读进环境变量，避免把密钥写进代码
from dotenv import load_dotenv
# 从 IPython.display 导入展示工具：在笔记本里漂亮显示 Markdown（本格先导入，后格可用）
from IPython.display import Markdown, display
# 从 openai 导入 OpenAI 客户端类：调用云端 Chat Completions API
from openai import OpenAI
```

```python
# ========== 环境：加载密钥并做格式自检 ==========

# 加载 .env；override=True 表示用文件里的值覆盖进程里已有同名环境变量
load_dotenv(override=True)
# 从环境变量读取 OpenAI 密钥（名字必须是 OPENAI_API_KEY）
api_key = os.getenv('OPENAI_API_KEY')

# 分支检查：缺密钥 / 前缀不对 / 首尾空白，分别给出英文排错提示（文案影响排查流程，保持英文）
if not api_key:
    # 完全没读到密钥
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    # 有值但不是常见的 sk-proj- 前缀，可能拷错了 Key
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    # strip 后变短说明首尾有空格/制表符，请求时常会鉴权失败
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    # 基本体检通过（仍不保证额度/权限一定可用）
    print("API key found and looks good so far!")
```

```python
# ========== 构造 messages 并调用 Chat Completions ==========

# 步骤 1：写 system / user 提示词（发给模型的英文 prompt 保持原样，勿翻译）
# system_prompt：设定「资深软件工程师」人设，负责给成长建议
system_prompt = """
                You are a senior software engineer with 25+ years
                of experience. You have to guide people on what 
                should they do to become a better engineer.
                """

# user_prompt：用户侧前缀；后面会拼上具体兴趣领域
user_prompt = """
              Here is the field of interest of the person : 
              """

# prompt：本次要咨询的领域字符串（可改成别的方向再跑）
prompt = "AI Engineer"

# 步骤 2：拼成 OpenAI messages 列表：system 定角色，user = 前缀 + 领域
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt + prompt}
]

# 步骤 3：创建客户端（默认从环境变量读 OPENAI_API_KEY）并发起非流式补全
openai = OpenAI()
# model / messages 参数保持原样；响应在 response 对象里
response = openai.chat.completions.create(model="gpt-5-nano", messages=messages)

# 步骤 4：取出第一条 choice 的助手正文并打印
print(response.choices[0].message.content)
```
