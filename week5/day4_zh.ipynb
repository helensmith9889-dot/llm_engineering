```python
# 查看第一条用例的字段：问题、类别、参考答案、用于检验检索的关键词

example = tests[0]
print(example.question)
print(example.category)
print(example.reference_answer)
print(example.keywords)
```

```python
# 导入两项评测：检索质量 evaluate_retrieval，与最终答案质量 evaluate_answer

from evaluation.eval import evaluate_retrieval, evaluate_answer
```

```python
# 评测检索：关键词是否出现在取回的 chunk 里（命中率）

evaluate_retrieval(example)
```

```python
# 查看评测结果对象（含 accuracy / completeness / relevance 等）

eval
```

```python
# 打印文字反馈与三项分数：准确性、完整性、相关性

print(eval.feedback)
print(eval.accuracy)
print(eval.completeness)
print(eval.relevance)
```
