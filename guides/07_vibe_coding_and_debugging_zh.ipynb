# Vibe coding 与调试

"Vibe coding" 是在 LLM 协助下写代码的亲切说法。用 Cursor 做尤其轻松美妙！但要做得好，有一些技巧——课程里会讲到。这里先给出几条关键提示：

1. 如果你在提示 ChatGPT 或 Claude 这类 LLM 写代码，请在提示中包含当前日期，并指示 LLM 生成使用最新版本 API 的代码。这对 OpenAI API 尤其重要！ChatGPT 经常试图使用它自己 API 的过时版本……

2. 在提示中也请让 LLM 把代码写得尽量短、尽量简单。LLM 似乎很爱加很多多余、不必要的检查，让代码变乱，也更难发现问题。

3. 把同一个问题问多个 LLM，选最清晰、最简单的那个答案。

4. 类似地：把一个 LLM 的答案拿给另一个 LLM，请它验证正确性，以及能否更简单或更清晰。

5. 请 LLM 给出解决某个问题的 3 种变体，选最清晰的那一个。

## 更大项目中的 Vibe Coding

尽量避免让 LLM 一次生成 100–200 行或更多代码；那样会非常难调试、难搞清哪里出了问题（除非你已经很有经验）。

相反：先把问题拆成小的、可独立测试的步骤，每一步都相对较小。如果你不确定如何拆解——这恰恰可以请 LLM 来做！

然后对每一个构建块：
- 用上面的技巧让 LLM 编写代码
- 也让 LLM 写测试来验证代码
- 自己测试，并确认它确实正确工作

这样你就能有信心地构建更大的项目。
