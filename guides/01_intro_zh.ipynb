# 欢迎来到 Guides！

## 简介

我设计这门课程时，希望它能适合各种背景的学员。如果你相对较新，我准备了一些技术简报来帮助你建立基础。

这些内容适合自学；你可以按自己的节奏推进、探索和实验。

我大量借助了 AI 伙伴来撰写部分指南，并尽量把它们组织得对课程尽可能有用、贴切。

课程只有一个要求：足够的耐心！请记住，最好的学习方式之一就是解决问题——如果你在难题前感到挫败，请记住：这正是学习发生的地方！如果我能帮上忙，请随时联系。
