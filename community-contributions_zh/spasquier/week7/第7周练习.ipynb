# 第 7 周练习 - 使用精简数据集微调开源模型

## 练习目标（理念）

本练习对应课程 **Week 7：微调（Fine-tuning）开源模型**。作者把完整训练与评估流程放在 **Google Colab** 上跑（GPU 更方便），本仓库笔记本只作**入口与索引**：记录训练笔记本与结果评估笔记本的链接。

你会对照理解：

- **精简数据集（slim / curated dataset）**：用更小、更干净的数据做监督微调，成本更低、迭代更快
- **微调训练笔记本**：装依赖、加载基座模型、准备数据、跑训练（常见会涉及 LoRA / QLoRA 等参数高效微调）
- **结果与评估笔记本**：对微调后的模型做推理对比、指标或人工抽检，验证是否比基座更好

## 和本课 Week 7 的关系

| 本课概念 | 本练习里你会看到 |
|----------|------------------|
| 开源模型微调 | Colab 训练笔记本中的训练循环与 checkpoint |
