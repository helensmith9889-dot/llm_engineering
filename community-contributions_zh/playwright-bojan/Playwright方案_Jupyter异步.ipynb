# Playwright 抓取演示（在 Jupyter 里跑异步）

本笔记本演示：如何借助 `nest_asyncio`，在 **JupyterLab / Notebook** 里运行基于 **异步 Playwright** 的抓取代码。

## 和本课概念的关系

| 概念 | 本练习里你会看到 |
|------|------------------|
| 异步 I/O `asyncio` | `asyncio.run(analyze_content())` 启动协程 |
| Jupyter 事件循环冲突 | `nest_asyncio.apply()` 允许在已有 loop 里再 `run` |
| Playwright 抓取 | 封装在同目录 `openai_scraper_playwright.py` |
| 结构化分析 | `analyze_content()` 返回可打印的分析结果 |

## 怎么跑

1. 确保同目录有 `openai_scraper_playwright.py`（以及它依赖的 Playwright / 密钥配置）
2. 从上到下依次运行代码格
3. 若第 3 格打印出结构化分析，说明异步路径在 Jupyter 里已打通


```python
# ========== 修补 Jupyter 事件循环：允许在 Notebook 里 asyncio.run ==========

# 导入 nest_asyncio：解决「Notebook 已有事件循环，不能再 asyncio.run」的冲突
import nest_asyncio
# 导入标准库 asyncio：运行协程（coroutine）、调度异步任务
import asyncio
# apply()：给当前事件循环打补丁，之后可以安全地再调用 asyncio.run(...)
nest_asyncio.apply()
```

```python
# ========== 导入同目录抓取模块：真正干活的逻辑在 py 文件里 ==========

# EnhancedOpenAIScraper：增强版 Playwright + OpenAI 抓取类（本格只导入，未必直接实例化）
# analyze_content：异步入口，内部完成抓取与内容分析，返回结构化结果
from openai_scraper_playwright import EnhancedOpenAIScraper, analyze_content
```

```python
# ========== 在 Jupyter 里启动异步分析：asyncio.run + nest_asyncio ==========

# asyncio.run：创建/复用事件循环，跑完 analyze_content() 协程，拿到返回值
result = asyncio.run(analyze_content())
# 打印分析结果（通常是结构化摘要 / dict / 字符串，取决于模块实现）
print(result)
```

若上方单元格已经打印出**结构化分析**，说明异步 Playwright 代码在 Jupyter 里运行成功。
