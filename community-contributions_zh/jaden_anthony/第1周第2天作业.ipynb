# 第 1 周第 2 天作业 —— 网站摘要器（Ollama）

## 练习目标（理念）

通过 **Ollama** 拉取并调用任意**开源本地模型**，对指定网站内容做分析 / 摘要。

## 和本课 Day 2 的关系

| 本课概念 | 本练习里你会看到 |
|----------|------------------|
| 本地开源模型 | `ollama pull` + `ollama.chat` |
| 网页抓取 | 同目录 `scraper.fetch_website_contents` |
| `messages`（system / user） | 讽刺风 system + 网站正文 user |

## 怎么跑

1. 本机已安装并启动 Ollama
2. 先跑「pull 模型」单元格，再依次运行导入 → 抓取 → 提示 → 消息 → 调用
3. 可改 `fetch_website_contents(...)` 的 URL，或改 system/user 提示风格后重跑


```python
# ========== 拉取本地模型：第一次跑需要下载权重 ==========

# 用 notebook 魔法命令调用 shell：让 Ollama 下载（或确认已有）llama3.2:latest
# 模型名字符串必须和后面 chat(model=...) 一致
!ollama pull llama3.2:latest
```

```python
# ========== 导入依赖：Ollama SDK + 本地抓取函数 ==========

# 从 ollama 包导入 chat：发起一轮本地聊天补全（Chat Completions 的本地版）
from ollama import chat
# 导入 ChatResponse 类型：给 response 变量做类型标注（Type Hint），方便阅读
from ollama import ChatResponse
# 从同目录 scraper 模块导入抓取函数：负责 HTTP + 清洗，返回网站正文文本
from scraper import fetch_website_contents
```

```python
# ========== 抓取：把目标站点正文读进变量 website ==========

# 调用抓取函数；URL 保持英文原样（决定抓哪一站）
# 返回值通常是长字符串，后面会拼进 user prompt
website = fetch_website_contents("https://anthropic.com")
```

```python
# ========== 提示词：system 定人设，user 放任务前缀 ==========

# system prompt：角色与语气（讽刺、不正经）；发给模型的指令保留英文
system_prompt = """You are non serious assistant who is very sarcastic and you specialise in website analyzation."""
# user prompt 前缀：后面会拼接 website 正文；字符串本身保留英文
user_prompt = """Here are website contents , please analyze -"""
```

```python
# ========== 组装 messages：API 需要的 role/content 列表 ==========

# 一条 system + 一条 user；user 内容 = 前缀 + 抓取正文
messages = [{'role':'system','content':system_prompt},{'role':'user','content':user_prompt + website}]
```

```python
# ========== 调用本地模型并打印回答 ==========

# chat(...)：同步调用 Ollama；model 名与 pull 的一致；messages 是对话上下文
# 类型标注 ChatResponse：提示 IDE / 读者这是 SDK 返回的结构化对象
response: ChatResponse = chat(model="llama3.2:latest", messages=messages)

# 取出助手回复文本并打印到笔记本输出区
print(response.message.content)
```
