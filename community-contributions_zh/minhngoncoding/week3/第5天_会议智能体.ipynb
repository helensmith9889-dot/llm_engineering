### 先热身：用可视化真正看清「模型推理」如何通过 OpenAI 工作

下面单元格会：

1. 用一句英文 `message` 当提示
2. 调用 `TokenPredictor` 拉取逐步 token 预测
3. 建成图 `G`，再用 `visualize_predictions` 画出来

目的不是改会议纪要逻辑，而是建立直觉：**生成 = 一步步选下一个 token**。


```python
# ========== Token 预测可视化：看清 next-token 推理 ==========
# 从本地 visualizer 模块导入：预测器、建图、绘图三个工具
from visualizer import TokenPredictor, create_token_graph, visualize_predictions

# 发给模型的用户消息（影响行为的英文 prompt，禁止翻译）
message = "In one sentence, describe the color orange to someone who has never been able to see"
# OpenAI 模型 id（字符串必须与账号可用模型一致，勿改）
model_name = "gpt-4.1-mini"

# 用指定模型创建预测器实例
predictor = TokenPredictor(model_name)
# 对 message 做逐步 token 预测，得到结构化 predictions
predictions = predictor.predict_tokens(message)
# 把预测结果转成图结构 G（供可视化）
G = create_token_graph(model_name, predictions)
# 画图；返回的 plt 通常是 matplotlib 相关对象
plt = visualize_predictions(G)
# 在笔记本中显示图像
plt.show()
```
