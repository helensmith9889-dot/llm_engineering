### 调试公式：用 10 行代码复现

出色的调试既是艺术也是科学。最好的掌握方式就是练习！但以下是要点：

1. Recreate（复现）  
稳定地复现你的问题

2. Simplify（简化）  
把问题缩减到最简单的复现方式。我说不清有多少人给我发来 100+ 行代码，说「Ed，这不行，请修一下」。事情不是这样的！我首先会做的，是把它缩到核心问题——理想情况下 10–20 行或更少。有一次，我向一位学生展示了如何用 2 行代码复现！

_关键是：这件事由**你**来做，比任何人做都容易得多。_ 所以虽然可能乏味，这几乎总是最好的起点。缩到几行代码。而且顺便说一句，当你这么做时，往往自己就搞清楚问题了！

3. Get help（寻求帮助）  
一旦你用几行代码复现了问题，如果仍看不出发生了什么——就该求助了！当然值得问问 ChatGPT 和朋友们；把短示例和完整 stack trace 给他们。

也可以联系我！我在这里是为了让你的学习体验尽可能棒；如果你卡住了，我会帮你解困。
