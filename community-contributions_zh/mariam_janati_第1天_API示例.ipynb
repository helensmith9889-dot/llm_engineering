# 第 1 天练习 —— 简单 API 解释生成器

## 练习目标（理念）

用 **OpenAI Chat Completions API**，配合 **system / user** 两条提示（Prompt），生成一段清晰、结构化的短解释。

这是 Day 1 的「最小可运行」示例：先写好角色与任务，再组装 `messages`，最后调用模型并打印回答。

## 和本课 Day 1 的关系

| 本课概念 | 本练习里你会看到 |
|----------|------------------|
| Environment Variables | `load_dotenv` 读 `.env`，避免把密钥写进代码 |
| `OpenAI()` 客户端 | 默认从环境变量取 API Key |
| system / user prompt | 定「怎么答」+「答什么」 |
| `temperature` | 控制随机性：偏低更稳，偏高更发散 |
| `chat.completions.create` | 真正发起一次对话补全 |

## 怎么跑

1. 配置好 `.env`（常见为 `OPENAI_API_KEY`）
2. 从上到下运行单元格
3. 可改 `user_prompt` 里的问题，或微调 `temperature` 观察回答变化

```python
# ========== 导入 + 环境：把工具箱和密钥准备好 ==========

# 导入标准库 os：需要时可读环境变量（Environment Variables）
import os
# 从 dotenv 导入 load_dotenv：把 .env 里的密钥读进进程环境，避免把密钥写进代码
from dotenv import load_dotenv
# 从 openai 导入 OpenAI 客户端类：后面用它调用 Chat Completions API
from openai import OpenAI

# 加载 .env；override=True 表示用文件里的值覆盖已有同名环境变量
load_dotenv(override=True)
# 创建客户端：未显式传 api_key 时，SDK 会自动读 OPENAI_API_KEY
openai = OpenAI()
```

```python
# ========== 第 1 步：写好 system / user 两条提示（Prompt）==========

# system：定助手的「人设与答题风格」；字符串保持英文（影响模型行为，不翻译）
system_prompt = "You are a helpful assistant that writes clear, structured, and concise answers."
# user：真正的用户问题；发给模型的内容保持原文
user_prompt = """
Explain what an API is in simple terms.
Give an example related to everyday life.
Keep it short.
"""

# ========== 第 2 步：组装 messages 列表（Chat Completions 的标准输入）==========

# 每条消息是 dict：role 标明身份，content 是文本内容
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]

# ========== 第 3 步：调用 OpenAI Chat Completions ==========
# temperature 控制随机性（Randomness）：
# 0.0-0.3 -> 更可预测、更一致（适合编码 / 严谨问答）
# 0.5-0.8 -> 平衡（常见默认区间）
# 0.9+ -> 更有创意，但也可能更不稳定

# 发起一次对话补全：指定模型、消息、温度
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages,
    temperature=0.7
)

# ========== 第 4 步：取出助手回复并打印 ==========
# choices[0] 是第一条候选；message.content 是助手生成的文本
print(response.choices[0].message.content)
```
