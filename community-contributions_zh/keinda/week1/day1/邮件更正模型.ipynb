# 邮件更正模型

## 练习目标（理念）

演示如何用 **OpenAI** 的系统提示（system prompt）与用户提示（user prompt），生成一封礼貌、专业的**更正邮件**。

典型场景：幻灯片日期写错、已发给教授 —— 请模型起草更正请求与简短主题行。


## 概述与本课关系

本笔记本调用 OpenAI 的 **Chat Completions API**：

| 本课概念 | 本练习里你会看到 |
|----------|------------------|
| `messages`（system / user） | system 定「会写专业邮件」，user 描述具体错误 |
| `chat.completions.create` | 一次非流式生成整封邮件 |
| 模型选择 | `gpt-4.1-nano`（轻量、便宜，适合短文案） |

模型被要求：撰写邮件，请教授更正此前提交的 PowerPoint 幻灯片上的错误日期（Fall 2025 → Fall 2024）。


```python
# ========== 用系统/用户提示生成礼貌的更正邮件 ==========

# 导入 openai：调用 Chat Completions API（模块级 openai.chat.completions.create）
import openai

# system prompt 保留英文：设定助手人设与写作专长，改译会改变语气/行为
system_prompt = "You are an AI assistant that specializes in writing and correcting professional emails."

# user prompt 保留英文：具体任务描述（错写成 Fall 2025，应改为 Fall 2024）
user_prompt = """
I accidentally wrote 'Fall 2025' instead of 'Fall 2024' on a slide in the PowerPoint presentation
I already sent to my professor. Please write a polite email asking my professor to correct the mistake,
and provide a short subject line for the email.
"""

# 步骤 2：构造 messages 列表（system + user）；这是 Chat Completions 的标准输入形状
messages = [
    {"role": "system", "content": system_prompt},
    {"role":"user", "content": user_prompt}
]

# 步骤 2：再次赋值 messages（原笔记本保留了重复写法；逻辑不改，仅加教学旁注）
# fill this in：课程模板里提醒学员自己填写 messages 的占位注释

messages = [{"role": "system", "content": system_prompt},
{"role":"user", "content": user_prompt}] # fill this in

# 步骤 3：调用 OpenAI Chat Completions；model 与 messages 决定生成内容
response = openai.chat.completions.create(model="gpt-4.1-nano", messages=messages)
# 步骤 4：打印助手回复正文（主题行 + 邮件正文通常都在 content 里）
print(response.choices[0].message.content)
```
