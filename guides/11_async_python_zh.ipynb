### 下面是一些示例：

````python
# 让我们定义一个 async 函数

import asyncio

async def do_some_work():
    print("Starting work")
    await asyncio.sleep(1)
    print("Work complete")
````

````python
# 这会做什么？

do_some_work()
````

````python
# 好，我们再试一次！

await do_some_work()
````

````python
# 这里有什么问题？

async def do_a_lot_of_work():
    do_some_work()
    do_some_work()
    do_some_work()

await do_a_lot_of_work()
````

````python
# 有趣的警告！我们来修复它

async def do_a_lot_of_work():
    await do_some_work()
    await do_some_work()
    await do_some_work()

await do_a_lot_of_work()
````

````python
# 现在让我们并行执行
# 重要的是要认识到：这并不是你可能习惯的那种「多线程」
# asyncio 库运行在单线程上，但它用循环在某个任务等待时在任务之间切换

async def do_a_lot_of_work_in_parallel():
    await asyncio.gather(do_some_work(), do_some_work(), do_some_work())

await do_a_lot_of_work_in_parallel()
````

### 最后——试着写一个调用 do_a_lot_of_work_in_parallel 的 Python 模块

见顶部链接；你需要在模块里写类似这样的代码：

```python
if __name__ == "__main__":
    asyncio.run(do_a_lot_of_work_in_parallel())
```
