# 解释「kernel」

这个 notebook 背后有一个 Python 进程，你每运行一个单元格它就执行该单元格。那个 Python 进程称为 Kernel。每个 notebook 有自己独立的 Kernel。

你可以点击上方的 "Restart Kernel" 按钮。

如果你随后尝试运行下一个单元格，会报错，因为 favorite_fruit 已不再定义。你需要从 notebook 顶部重新运行单元格。然后下一个单元格应该就能正常运行。

````python
# 若刚重启过 Kernel，请先从顶部重新运行前面的单元格，再运行本单元格

print(f"My favorite fruit is {favorite_fruit}")
````

# 单元格输出

当你执行一个单元格时，标准输出以及最后一条语句的结果会写到代码正下方的区域，称为「单元格输出」（cell output）。当你从文件菜单保存 Notebook（或 ctrl+S / command+S）时，输出也会被保存，这是记录发生了什么的有用方式。

你可以通过点击工具栏中的 "Clear All Outputs" 来清理。在把代码推送到像 GitHub 这样的仓库之前，最好清空输出，否则文件可能很大且更难阅读。

````python
spams = ["spam"] * 1000
print(spams)

# 运行后或许值得清空输出！
````

# 使用 markdown

那么这些有文字的区域（比如这个）是怎么回事？其实还有一种不同的单元格，叫做「Markdown」单元格，用来添加像这样的说明。点击 [+ Markdown] 按钮添加新的 markdown 单元格。

用 Markdown 格式添加一些注释，或许可以从这里复制粘贴：

```
# This is a heading
## This is a sub-head
### And a sub-sub-head

I like Jupyter Lab because it's
- Easy
- Flexible
- Satisfying
```

然后只需在单元格中按 Shift+Return，就会变成格式化文本。
若要删除，点击单元格并按垃圾桶图标。

# 课程中会遇到的一些小知识点

这不一定是 notebooks 的特性，但这个包很值得了解，在 notebooks 中很有用。

包 `tqdm` 会在你包装任何可迭代对象时打印漂亮的进度条。

````python
# 下面是没有进度条的代码
# 它会花 10 秒，而你在想到底发生了什么..

import time

spams = ["spam"] * 1000

for spam in spams:
    time.sleep(0.01)
````

````python
# 现在，带上一个漂亮的小进度条：

import time
from tqdm import tqdm

spams = ["spam"] * 1000

for spam in tqdm(spams):
    time.sleep(0.01)
````

````python
# 换个话题，这里有一种用 markdown 打印输出的实用方法

from IPython.display import Markdown, display

display(Markdown("# This is a big heading!\n\n- And this is a bullet-point\n- So is this\n- Me, too!"))
````
