## 使用代码进行标记

```python
import tiktoken

encoding = tiktoken.encoding_for_model("gpt-4.1-mini")

tokens = encoding.encode("Hi my name is Al-abass and I like banoffee pie")
```

```python
tokens
```

```python
for token_id in tokens:
    token_text = encoding.decode([token_id])
    print(f"{token_id} = {token_text}")
```

```python
import os
from dotenv import load_dotenv

load_dotenv(override=True)
api_key = os.getenv('GROQ_API_KEY')

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("gsk_"):
    print("An API key was found, but it doesn't start with gsk_; please check you're using the right Groq key")
else:
    print("API key found and looks good so far!")
```

### 您应该对下一个单元正在做的事情感到非常满意！

_我正在创建 OpenAI Python 客户端库的一个新实例，这是一个轻量级包装器，用于对端点进行 HTTP 调用以调用 GPT LLM 或其他 LLM 提供商_

```python
from openai import OpenAI

groq = OpenAI(base_url="https://api.groq.com/openai/v1", api_key=api_key)
```

### 发送给 OpenAI 的消息是一个字典列表

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Hi! I'm Ed!"}
    ]
```

```python
response = groq.chat.completions.create(model="llama-3.3-70b-versatile", messages=messages)
response.choices[0].message.content
```

### 好的，现在让我们问一个后续问题

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "What's my name?"}
    ]
```

```python
response = groq.chat.completions.create(model="llama-3.3-70b-versatile", messages=messages)
response.choices[0].message.content
```

### 等等，什么？

我们刚刚告诉过你！

这是怎么回事？？

事情是这样的：对 LLM 的每次调用都是完全无状态的。每一次都是一次全新的通话。作为人工智能工程师，我们的工作就是设计一些技术来给人留下法学硕士有“记忆”的印象。

```python
messages = [
    {"role": "system", "content": "You are a sassy assistant"},
    {"role": "user", "content": "Hi! I'm Al-abass!"},
    {"role": "assistant", "content": "Hi Van! How can I assist you today?"},
    {"role": "user", "content": "What's my name?"}
    ]
```

```python
response = groq.chat.completions.create(model="llama-3.3-70b-versatile", messages=messages)
response.choices[0].message.content
```

## 回顾一下

如果这对你来说是显而易见的，我深表歉意 - 但仍然有必要强调一下：

1. 对 LLM 的每次调用都是无状态的
2. 我们每次都会在输入提示中传递到目前为止的整个对话
3. 这给人一种 LLM 有记忆的错觉——它显然保留了对话的上下文
4. 但这是一个技巧；这是每次提供整个对话的副产品
5. LLM 只是预测序列中最有可能的下一个标记；如果该序列包含“我的名字是 Ed”，然后包含“我的名字是什么？”然后它会预测.. Ed！

ChatGPT 产品正是使用了这个技巧 - 每次您发送消息时，都会传入整个对话。

“这是否意味着我们每次都必须为迄今为止的所有对话支付额外费用”

确实如此。这就是我们想要的。我们希望法学硕士能够预测序列中的下一个标记，回顾整个对话。我们希望进行计算，因此我们需要为此支付电费！
