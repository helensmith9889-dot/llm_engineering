## 好，开始我们的第一个项目

下面定义 `Website` 类：输入 URL，尽量抽出 `title` 与可见正文 `text`，供后续塞进 prompt。


````python
# ========== Website 类：BS4 优先，失败/正文太少再回退 Selenium ==========

# 若不熟悉类，可看 "Intermediate Python" notebook

# 有些网站会拒掉「不像浏览器」的请求，所以带上常见 Chrome User-Agent
headers = {
 "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
}

class Website:
    def __init__(self, url):
        # 保存原始 URL
        self.url = url
        # 标题与正文先置空，后面两种抓取路径再填充
        self.title = None
        self.text = None

        # --- 路径 A：先试 BeautifulSoup（轻量、快）---
        try:
            # GET 网页；timeout=10 避免卡死
            response = requests.get(url, headers=headers, timeout=10)
            # 用 html.parser 解析响应字节
            soup = BeautifulSoup(response.content, 'html.parser')
            # 有 <title> 就取字符串，否则给占位英文（下游 prompt 会用到）
            self.title = soup.title.string if soup.title else "No title found"
            # 删掉 script/style/img/input，减少导航噪音与无关标签
            for irrelevant in soup.body(["script", "style", "img", "input"]):
                irrelevant.decompose()
            # 抽出可见文本：换行分隔并 strip
            self.text = soup.body.get_text(separator="\n", strip=True)
        except Exception:
            # 任何异常都先记成「没正文」，交给后面 Selenium 兜底
            self.text = None

        # --- 路径 B：若 BeautifulSoup 失败或正文太短（<100），再试 Selenium ---
        if not self.text or len(self.text) < 100:
            try:
                # 无头 Chrome 选项：不弹窗、容器友好
                options = Options()
                options.add_argument("--headless")
                options.add_argument("--no-sandbox")
                options.add_argument("--disable-dev-shm-usage")
                # 无需指定 executable_path；Selenium Manager 会处理驱动
                driver = webdriver.Chrome(options=options)
                # 打开目标 URL（会执行页面 JS）
                driver.get(url)
                # 等待 JS 渲染（简单粗暴的 sleep）
                time.sleep(3)  # 等待 JS 加载
                # 优先用浏览器 title；否则保留 BS4 已有 title；再否则占位
                self.title = driver.title or self.title or "No title found"
                try:
                    # 取 <body> 的可见文本
                    body = driver.find_element(By.TAG_NAME, "body")
                    self.text = body.text
                except Exception:
                    # 找不到 body 时的占位文案（保持英文原样）
                    self.text = "No body text found"
                # 用完务必退出浏览器进程
                driver.quit()
            except Exception as e:
                # Selenium 也失败：把异常信息写进 text，便于后面排查
                self.text = f"Selenium failed: {e}"
````

````python
# ========== 试抓取：可改 URL，并加 print 跟进中间结果 ==========

# 实例化 Website：内部会走 BS4，必要时再 Selenium
ed = Website("https://edwarddonner.com")
# 打印标题
print(ed.title)
# 打印正文（可能很长）
print(ed.text)
````

## 提示词类型

你可能已经知道——如果还不知道，很快就会非常熟悉！

像 GPT-4o 这类模型被训练成按特定方式接收指令。

它们期望收到：

**System 提示**：告诉模型要执行什么任务、用什么语气

**User 提示**——对话开场，模型需要回复的内容

下一格先写死 `system_prompt`；再下一格用函数根据 `Website` 拼 `user` 提示。


````python
# ========== system prompt：定「网站摘要助手」角色（英文指令不翻译）==========

# 定义 system 提示——稍后可实验，例如改成用西班牙语 Markdown 回复
# 发给模型的指令字符串必须保持英文原样：改译会改变回答风格/行为
system_prompt = "You are an assistant that analyzes the contents of a website \
and provides a short summary, ignoring text that might be navigation related. \
Respond in markdown."
````

````python
# ========== user prompt：把网站标题 + 正文拼进给模型的请求 ==========

# 编写用于请求网站摘要的 User 提示：
def user_prompt_for(website):
    # 先告诉模型「你正在看哪个标题的网站」
    user_prompt = f"You are looking at a website titled {website.title}"
    # 再说明任务：短摘要、含新闻/公告也要概括（英文任务句不翻译）
    user_prompt += "\nThe contents of this website is as follows; \
please provide a short summary of this website in markdown. \
If it includes news or announcements, then summarize these too.\n\n"
    # 最后附上抓取到的正文
    user_prompt += website.text
    return user_prompt
````

````python
# ========== 预览：看看拼出来的 user prompt 长什么样 ==========

# 对刚才抓到的 edwarddonner.com 网站对象拼 prompt 并打印
print(user_prompt_for(ed))
````

## 消息（Messages）

OpenAI 的 API 期望按特定结构接收消息。
许多其他 API 也采用相同结构：

```python
[
    {"role": "system", "content": "system message goes here"},
    {"role": "user", "content": "user message goes here"}
]
```

作为预览，接下来 2 格会做一次很简单的调用——我们还不会为难强大的 GPT（暂时！）


````python
# ========== 预览用 messages：故意用「毒舌助手」角色问 2+2 ==========

# role=system：定语气；role=user：具体问题（字符串保持英文）
messages = [
    {"role": "system", "content": "You are a snarky assistant"},
    {"role": "user", "content": "What is 2 + 2?"}
]
````

````python
# ========== 预览调用：把上面的 messages 发给 gpt-4o-mini ==========

# 预览：用 system 与 user 消息调用 OpenAI：
response = openai.chat.completions.create(model="gpt-4o-mini", messages=messages)
# 打印助手回复正文
print(response.choices[0].message.content)
````

## 现在用函数为 GPT-4o-mini 构建有用的 messages

把前面的 `system_prompt` 与 `user_prompt_for(website)` 组装成 API 期望的列表结构。


````python
# ========== messages_for：精确构造 system + user 两条消息 ==========

# 看此函数如何精确构造上面的格式
def messages_for(website):
    return [
        # system：全局任务与语气
        {"role": "system", "content": system_prompt},
        # user：该网站的标题 + 正文 + 摘要请求
        {"role": "user", "content": user_prompt_for(website)}
    ]
````

````python
# ========== 试一下：看 messages 结构；也可换几个网站对象 ==========

# 对 ed 网站对象组装 messages（返回 list，笔记本会展示）
messages_for(ed)
````

## 组合起来——OpenAI 的 API 非常简单！

抓取 → 拼 messages → `chat.completions.create` → 取出 `content`，整条链路就通了。


````python
# ========== summarize：URL 进，摘要字符串出 ==========

# 现在：调用 OpenAI API。你很快会非常熟悉！
def summarize(url):
    # 先抓取网页，得到 title/text
    website = Website(url)
    # 再把 messages_for(website) 发给 gpt-4o-mini
    response = openai.chat.completions.create(
        model = "gpt-4o-mini",
        messages = messages_for(website)
    )
    # 返回模型生成的 Markdown 摘要正文
    return response.choices[0].message.content
````

````python
# ========== 端到端试跑：摘要 edwarddonner.com ==========

# 直接调用 summarize；笔记本会显示返回的字符串
summarize("https://edwarddonner.com")
````

````python
# ========== display_summary：用 Markdown 在 Jupyter 里漂亮展示 ==========

# 用 markdown 在 Jupyter 输出中漂亮展示
def display_summary(url):
    # 先拿到摘要字符串
    summary = summarize(url)
    # 再渲染成 Markdown 单元格输出
    display(Markdown(summary))
````

````python
# ========== 漂亮展示：同一 URL，看渲染后的摘要 ==========

display_summary("https://edwarddonner.com")
````

````python
# ========== 自己的变体示例：英超解说说进球（四步走）==========

# 第 1 步：编写提示词（发给模型的英文指令/事实不翻译）
system_prompt = "You are a soccer commentator for premier league games mimicing Peter Dury vocabulary and phrases to serenade a player on scoring a goal"
user_prompt = """
    Dominik Szoboslai of Liverpool just hit a 32 yard free kick in the 82nd min against Arsenal, making the score 1 nil to liverpool and closer to winnning the game.
"""

# 第 2 步：组装 messages 列表（system + user）
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]

# 第 3 步：调用 OpenAI Chat Completions
response = openai.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages
)

# 第 4 步：打印结果正文
print(response.choices[0].message.content)
````
