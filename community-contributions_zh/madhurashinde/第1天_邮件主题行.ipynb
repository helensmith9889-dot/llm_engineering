## 为邮件建议合适的主题行

## 练习目标（理念）

做一个最小可用的 **邮件主题行（subject line）生成器**：

- **输入**：一封邮件正文（user prompt）
- **输出**：模型建议的主题行字符串
- **后端**：OpenAI Chat Completions（本练习用 `gpt-4.1-mini`）

## 和本课 Day 1 的关系

| 本课概念 | 本练习里你会看到 |
|----------|------------------|
| `.env` + `OPENAI_API_KEY` | `load_dotenv` + 前缀 / 空白校验 |
| `OpenAI()` 客户端 | 默认从环境读密钥 |
| `messages`（system / user） | system 定任务，user 放邮件正文 |
| `chat.completions.create` | 一次调用拿到主题行建议 |

## 怎么跑

1. 在项目根或当前环境准备好 `.env`（含 `OPENAI_API_KEY`）
2. 从上到下运行单元格
3. 把 `user_prompt` 换成你自己的邮件正文，再跑最后一格看新主题行


```python
# ========== 导入：环境、展示与 OpenAI 客户端 ==========

# os：通过环境变量读取 OPENAI_API_KEY
import os
# load_dotenv：把 .env 文件加载进进程环境，避免把密钥写进笔记本
from dotenv import load_dotenv
# Markdown / display：本练习最后用 print；导入保留以便你改成漂亮展示
from IPython.display import Markdown, display
# OpenAI：Chat Completions 的官方 Python SDK 入口
from openai import OpenAI
```

```python
# ========== 加载并校验 API Key（常见踩坑自检） ==========

# override=True：以 .env 文件为准，覆盖进程里已有同名变量
load_dotenv(override=True)
# 从环境读取 OpenAI 密钥
api_key = os.getenv('OPENAI_API_KEY')

# 分支 1：完全没读到密钥 —— 提示去排查笔记本（文案保持英文）
if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
# 分支 2：有值但不是课程常用的 sk-proj- 前缀
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
# 分支 3：首尾有空格 / Tab，复制粘贴时常见
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
# 分支 4：看起来可用
else:
    print("API key found and looks good so far!")
```

```python
# ========== 创建 OpenAI 客户端 ==========

# 无参构造：默认从环境变量 OPENAI_API_KEY 取密钥（上一格已校验）
openai = OpenAI()
```

```python
# ========== 提示词：system 定任务，user 放示例邮件正文 ==========

# system：说明助手职责是「为邮件提供合适主题行」；字符串保持英文
system_prompt = "You are an assistant that provides subject line appropriate for the email"
# user：一封招聘回复示例（Snowflake Associate Solution Engineer）；可改成你的真实邮件
user_prompt = """
    Hi Anna, 

Thank you for your interest in pursuing a career with Snowflake. We’ve received your application for the Associate Solution Engineer role.

If your application meets the requirements for the role, we will contact you regarding the next steps. 
"""
```

```python
# ========== 调用 Chat Completions，打印主题行建议 ==========

# messages：第一轮 system，第二轮 user（标准双角色对话）
messages = [{"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}] 

# model id 保持原样 gpt-4.1-mini；一次非流式调用
response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages
    )
# 取出助手回复正文（期望是主题行或主题行建议）
res = response.choices[0].message.content

# 打印到笔记本输出区
print(res)
```
