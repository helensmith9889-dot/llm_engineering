````python
# ========== 工具函数：从（可能加密的）PDF 抽取文本 ==========

# 定义函数：输入 PDF 路径与密码，返回全文；失败时打印错误并返回 None
def extract_text_from_pdf(pdf_path, pdf_password):
    """
    Extracts text from encrypted or unencrypted PDF files.
    If encrypted, decrypts it temporarily using pikepdf.
    """
    # 提示正在读取哪个文件（路径保持原样）
    print(f"Reading PDF from: {pdf_path}...")

    # 没有密码就无法解密：提示去 .env 配置 PDF_PASSWORD（示例文案保持英文）
    if not pdf_password:
        print(f"--- ERROR ---")
        print(f"No 'PDF_PASSWORD' found in your .env file.")
        print("Please add your PAN to the .env file (e.g., PDF_PASSWORD=\"ABCDE1234F\")")
        return None

    try:
        # 临时解密文件路径：在原文件名后加 _decrypted.pdf
        temp_path = os.path.splitext(pdf_path)[0] + "_decrypted.pdf"

        # 用 pikepdf 打开并保存一份已解密副本（可处理 AES 等）
        print("Attempting decryption using pikepdf...")
        with pikepdf.open(pdf_path, password=pdf_password) as pdf:
            pdf.save(temp_path)
        print("Decryption successful!")

        # 再用 pypdf 逐页抽取文本
        reader = PdfReader(temp_path)
        full_text = ""
        for page in reader.pages:
            # extract_text：取出该页可见文本，并在页之间加换行
            full_text += page.extract_text() + "\n"

        print("✅ PDF text extracted successfully.")
        # 删掉临时解密文件，避免磁盘残留敏感内容
        os.remove(temp_path)  # cleanup
        return full_text

    except WrongPasswordError:
        # 注意：WrongPasswordError 需由 pikepdf 提供；此处逻辑保持原样不改 import
        print(f"--- ERROR --- Wrong PDF password. Check your PAN in the .env file.")
        return None
    except Exception as e:
        # 其它异常：打印信息后返回 None，让后续步骤可判断失败
        print(f"An error occurred while reading the PDF: {e}")
        return None
````

````python
# ========== 环境：加载 .env，并检查 OpenAI API Key ==========

# 加载名为 .env 的文件中的环境变量；override=True 表示覆盖已有同名变量
load_dotenv(override=True)
# 读取 OpenAI 密钥
api_key = os.getenv('OPENAI_API_KEY')
# 读取 PDF 密码（本练习扩展：Zerodha 合同确认单常用 PAN 作密码）
pdf_password = os.getenv('PDF_PASSWORD')
# 读取 PDF 文件路径
pdf_file_path = os.getenv("PDF_PATH")

# 检查密钥形态（提示文案保持英文原样）

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")
````

## 提示词的类型

你可能已经知道——如果还不知道，接下来你会非常熟悉！

像 GPT 这样的模型，被训练成以特定方式接收指令。

它们期望收到：

**系统提示词（system prompt）**：告诉它们要执行什么任务，以及应使用什么语气

**用户提示词（user prompt）**——它们应回复的对话开场白


````python
# ========== 定义 system prompt：规定「怎么总结网站」 ==========

# 定义我们的系统提示词——稍后可以实验，例如把最后一句改成 'Respond in markdown in Spanish.'

# 发给模型的指令保持英文原样（会影响语气与输出格式）
system_prompt = """
You are a funny assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
````

````python
# ========== 定义 user prompt 前缀：后面会拼上网页正文 ==========

# 定义我们的用户提示词前缀（真正的网页正文会在 messages_for 里拼接）

user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""
````

## 消息（messages）

OpenAI 的 API 期望以特定结构接收消息。
许多其他 API 也共享这种结构：

```python
[
    {"role": "system", "content": "system message goes here"},
    {"role": "user", "content": "user message goes here"}
]
```
先给你一个预览：接下来两个单元格会做一个相当简单的调用——我们（暂时）不会为难强大的 GPT！


````python
# ========== 再练一次 messages 结构，并立刻调用模型 ==========

# 构造两条消息：system 定人设，user 提问题（字符串保持英文原样）
messages = [
    {"role": "system", "content": "You are a strict militay officer assistant"},
    {"role": "user", "content": "What is 2 + 2?"}
]

# 用较小的模型做一次快速调用，观察 content
response = openai.chat.completions.create(model="gpt-4.1-nano", messages=messages)
response.choices[0].message.content
````

## 现在用函数为 GPT-4.1-mini 构建有用的 messages


````python
# ========== 封装 messages_for：把「网页正文」变成标准 messages ==========

# 看这个函数如何精确创建上面的格式

def messages_for(website):
    # 返回 [system, user]：user 的 content = 前缀说明 + 网页正文
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
````

````python
# ========== 试一下 messages_for：看看拼出来的结构 ==========

# 试试看，然后再试几个网站

messages_for(ed)
````

## 是时候整合了——OpenAI 的 API 非常简单！


````python
# ========== 整合：URL → 抓取 → Chat Completions → 摘要文本 ==========

# 现在：调用 OpenAI API。你会对它非常熟悉！

def summarize(url):
    # 先抓取网页正文
    website = fetch_website_contents(url)
    # 再把正文放进 messages，调用云端模型生成摘要
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(website)
    )
    # 只返回助手文本，方便后续 display
    return response.choices[0].message.content
````

````python
# ========== 调用 summarize：对示例网站生成摘要 ==========

summarize("https://edwarddonner.com")
````

````python
# ========== 包装 display_summary：用 Markdown 漂亮展示 ==========

# 一个函数，用 markdown 在输出中漂亮地显示

def display_summary(url):
    # 先拿到摘要字符串
    summary = summarize(url)
    # 再渲染为 Markdown
    display(Markdown(summary))
````

````python
# ========== 端到端试跑：抓取 + 总结 + Markdown 展示 ==========

display_summary("https://edwarddonner.com")
````

````python
# ========== 学生扩展：Zerodha 合同确认单 → 交易摘要 ==========

# 步骤 1：创建提示词（system prompt 保持英文原样，会直接影响抽取字段与格式）

system_prompt = """
You are a specialized financial assistant designed to analyze and summarize daily trading reports from Zerodha. 
Your primary function is to extract key financial figures from the text of a contract note provided by the user and present them in a clear, 
concise, and structured summary.

Your response MUST include the following, calculated from the provided text:
1.  **Total Turnover**: The sum of all buy and sell transaction values.
2.  **Itemized Charges**: A breakdown of all taxes and fees. This includes Brokerage, STT (Securities Transaction Tax), 
Exchange Transaction Charges, GST, SEBI Turnover Fees, and Stamp Duty.
3.  **Total Charges**: The sum of all the itemized charges.
4.  **Net Realized Profit or Loss**: The final profit or loss after all charges have been deducted. Clearly label it as 'Profit' or 'Loss'.

Respond ONLY in Markdown format. Use headings and bolding for clarity. Do not include any conversational phrases, greetings, or explanations.
"""

# 用前面的 PDF 工具，把合同确认单正文抽出来当作 user_prompt
user_prompt = extract_text_from_pdf(pdf_file_path, pdf_password)

# 步骤 2：组装 messages 列表（仅当成功抽到文本时才继续）
if user_prompt:
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ] # fill this in

    # 步骤 3：调用 OpenAI
    print("Connecting to OpenAI to get your summary...")
    try:
        # 单独再创建一个客户端实例（与前面的 openai 变量并存，逻辑保持原样）
        openai_client = OpenAI()
        response = openai_client.chat.completions.create(
            model="gpt-4o-mini",  # gpt-4o is excellent for this
            messages=messages
        )

        # 步骤 4：打印结果（标题与错误文案保持英文原样）
        print("\n--- 📈 Your Trading Summary ---")
        print(response.choices[0].message.content)

    except Exception as e:
        print(f"--- ERROR ---")
        print(f"An error occurred while contacting OpenAI: {e}")
````
