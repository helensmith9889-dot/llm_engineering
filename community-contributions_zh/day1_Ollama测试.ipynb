```python
# ========== 拉取本地模型 + 用 OpenAI 兼容接口调用 Ollama ==========
# openai = OpenAI()  # 云端写法（此处改用本地 Ollama）
# 本机只需执行一次拉取模型（shell magic：以 ! 开头，交给系统 shell 执行）
# llama3.2：模型名字符串必须与本机 `ollama list` 里的名字一致
!ollama pull llama3.2

# 从 openai 包导入 OpenAI 客户端类：本格用它打「OpenAI 兼容」的本地端点，而不是云端
from openai import OpenAI
# MODEL：集中写模型 id；后面 create(...) 只引用常量，改模型时只改这一处
MODEL = "llama3.2"
# 创建客户端：base_url 指向本机 Ollama 的 OpenAI 兼容层 /v1；api_key 对本地通常任意非空即可
openai = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")

# chat.completions.create：Chat Completions API；messages 是 system/user 角色列表
response = openai.chat.completions.create(
 model=MODEL,
 messages=[{"role": "system", "content": "you are a wierd assistant"},{"role": "user", "content": "What is 2 + 2?"}]
)

# print(response.choices[0].message.content)  # 也可只打印回复正文
# 打印完整 response 对象：便于观察 choices、model、usage 等字段结构（教学用）
print(response)
```
