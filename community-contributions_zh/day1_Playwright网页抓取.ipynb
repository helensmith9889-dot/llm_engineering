## 提示词类型

你可能已经知道——如果还不知道，很快就会非常熟悉！

像 GPT 这类模型，训练时就约定了「按特定方式接收指令」。

它们期望收到：

**System 提示**：告诉模型要执行什么任务、用什么语气

**User 提示**——对话开场，模型需要回复的内容


````python
# ========== 定义 system 提示：角色 + 输出格式（Markdown） ==========
# 定义 system 提示——稍后可实验，例如改成用西班牙语 Markdown 回复
# 三引号字符串保持英文：这是发给模型的指令，翻译会改变语气/行为
system_prompt = """
You are a snarkyassistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
````

````python
# ========== 定义 user 提示前缀：后面会拼上网页正文 ==========
# 定义 user 提示
# 英文指令原样保留；真正的网站内容在 messages_for() 里追加
user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""
````

## 消息（Messages）

OpenAI 的 API 期望按特定结构接收消息。
许多其他 API 也采用相同结构：

```python
[
    {"role": "system", "content": "system message goes here"},
    {"role": "user", "content": "user message goes here"}
]
```

作为预览，接下来 2 格会做一次很简单的调用——我们还不会为难强大的 GPT（暂时！）


````python
# ========== 最小 messages 示例：system + user，再调用 gpt-4.1-nano ==========
# 手写两条消息：角色分工一目了然
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "What is 2 + 2?"}
]

# 用较小的 nano 模型试水（model id 原样保留）
response = openai.chat.completions.create(model="gpt-4.1-nano", messages=messages)
# 展示回复正文
response.choices[0].message.content
````

## 现在用函数为 GPT-4.1-mini 构建有用的 messages


````python
# ========== 把 system_prompt + 网页正文 组装成标准 messages ==========
# 看此函数如何精确构造上面的格式
# website：已抓取的页面文本（或标题+正文）
def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        # user：前缀说明任务 + 网页内容
        {"role": "user", "content": user_prompt_prefix + website}
    ]
````

````python
# ========== 试跑 messages_for：看看发给 API 的完整结构 ==========
# 试一下，再换几个网站
# 传入上一格抓到的 ed；Jupyter 会展示返回的 list[dict]
messages_for(ed)
````

## 组合起来——OpenAI 的 API 非常简单！


````python
# ========== 端到端：URL → 抓取 → Chat Completions → 摘要字符串 ==========
# 现在：调用 OpenAI API。你很快会非常熟悉！

def summarize(url):
    # 1) 抓网页文本
    website = fetch_website_contents(url)
    # 2) 用 gpt-4.1-mini + messages_for 生成摘要
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(website)
    )
    # 3) 只返回正文
    return response.choices[0].message.content
````

````python
# ========== 对 edwarddonner.com 跑一遍 summarize ==========
summarize("https://edwarddonner.com")
````

````python
# ========== 用 IPython Markdown 漂亮展示摘要 ==========
# 用 markdown 漂亮展示输出

def display_summary(url):
    # 先拿到摘要字符串
    summary = summarize(url)
    # 在笔记本里渲染为 Markdown（而不是纯文本转义）
    display(Markdown(summary))
````

````python
# ========== 展示 edwarddonner.com 的 Markdown 摘要 ==========
display_summary("https://edwarddonner.com")
````

**为什么对 JavaScript 网站要用 Playwright？**

像 openai.com 这类现代站点会用 React、Next.js 等框架，在浏览器里动态渲染内容。`requests` + `BeautifulSoup` 只能拿到最初的 HTML，看不到 JS 渲染后的正文。

Playwright 的做法是启动真实浏览器（Chromium / Firefox / WebKit），执行 JavaScript，并等待页面渲染完成，从而抓到完整内容。相比 Selenium，它更现代、异步支持更好、跨浏览器也更方便。

只要遇到提示 “Enable JavaScript”，或用普通 HTTP 抓取结果是空的页面，就很适合改用 Playwright。


````python
# ========== Playwright 异步抓取：启动 Chromium，取正文/标题/链接等 ==========
# 目标站点：大量 JS 渲染，简单 requests 常失败
url = "https://openai.com"
# asyncio：异步事件循环工具（本格主要靠 notebook 的 top-level await）
import asyncio
# async_playwright：异步 Playwright 入口；Playwright 类型标注
from playwright.async_api import Playwright, async_playwright
# sync 版一并导入（原 notebook 保留；本格实际走 async）
from playwright.sync_api import Playwright, sync_playwright

async def run(playwright: Playwright, url: str):
    # Launch Chromium in non-headless mode (visible UI)
    # headless=False：有界面，便于观察加载过程（CI 里通常改 True）
    browser = await playwright.chromium.launch(headless=False)
    # 新页面并设视口大小
    page = await browser.new_page(viewport={"width": 1600, "height": 900})
    
    # 导航到 URL，等待加载
    await page.goto(url)
    
    # Get title：浏览器标题栏文本
    title = await page.title()
    
    # Get all text content：<body> 可见文本
    body_text = await page.inner_text('body')
    
    # Get all headings：所有 h1/h2/h3 文本列表
    headings = await page.locator('h1, h2, h3').all_text_contents()
    
    # Get all links：在浏览器里用 JS 把 <a> 映射成 {text, href}
    links = await page.locator('a').evaluate_all(
        "elements => elements.map(e => ({text: e.textContent, href: e.href}))"
    )
    
    # Get full HTML：完整页面 HTML 字符串
    html = await page.content()

     # Get structured data using evaluate：在页面上下文执行一段 JS，返回结构化字典
    data = await page.evaluate('''() => {
        return {
            headings: Array.from(document.querySelectorAll('h1, h2, h3')).map(h => h.textContent),
            paragraphs: Array.from(document.querySelectorAll('p')).map(p => p.textContent),
            links: Array.from(document.querySelectorAll('a')).map(a => ({
                text: a.textContent,
                href: a.href
            }))
        }
    }''')
    
    # 关闭浏览器，释放资源
    await browser.close()
    
    # 打包结果；html 默认不返回（需要时取消注释）
    return {
        "url": url,
        "title": title,
        "body_text": body_text,
        "headings": headings,
        "links": links,
        "structured_data": data,
        # "html": html  # Uncomment if you need raw HTML
    }

# async with：进入异步 Playwright 上下文，结束后自动清理
async with async_playwright() as playwright:
    # 跑抓取协程，拿到字典结果
    result = await run(playwright, url=url)
    # 先打印正文，确认 JS 渲染后的内容确实抓到了
    print(result['body_text'])
````

````python
# ========== 智能抓取 + 摘要：普通 HTML 走 requests，JS 站走 Playwright ==========
# BeautifulSoup：解析 HTML；requests：发 HTTP GET
from bs4 import BeautifulSoup
import requests

# 抓取网站用的标准请求头（伪装成常见桌面 Chrome，降低被拒概率）
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
}

async def get_js_web_content(url):
    # 复用上一格的 async run()：用 Playwright 取 body_text
    async with async_playwright() as playwright:
        result = await run(playwright, url=url)
        return result['body_text']


async def fetch_website_contents_new(url):
    """
    Return the title and contents of the website at the given url;
    truncate to 2,000 characters as a sensible limit
    """
    # 先尝试普通 HTTP：快、轻量
    response = requests.get(url, headers=headers)
    # 用 html.parser 建 DOM 树
    soup = BeautifulSoup(response.content, "html.parser")
    # 有 <title> 就取字符串，否则占位
    title = soup.title.string if soup.title else "No title found"
    
    # print(soup.body.get_text().lower()) 
    # 若页面提示需要启用 JS：改走 Playwright 异步抓取
    if "enable javascript" in soup.body.get_text().lower():
        response = await get_js_web_content(url)
        return response
    
    if soup.body:
        # 去掉 script/style/img/input 等对摘要无用的节点
        for irrelevant in soup.body(["script", "style", "img", "input"]):
            irrelevant.decompose()
        # 取出可见文本，换行分隔并 strip
        text = soup.body.get_text(separator="\n", strip=True)
    else:
        text = ""
    # 标题 + 正文，截断到 2000 字符，控制 prompt 长度/费用
    return (title + "\n\n" + text)[:2_000]

async def summarize_new(url):
    # 异步抓取（可能内部再 await Playwright）
    website = await fetch_website_contents_new(url)
    # 先打印原始抓取结果，便于对照摘要
    print(website)
    # 仍用同一套 messages_for + gpt-4.1-mini
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(website)
    )
    return response.choices[0].message.content

# 用 markdown 漂亮展示输出

async def display_summary_new(url):
    # await 异步摘要
    summary = await summarize_new(url)
    # 调试打印摘要字符串（原文保留）
    print('Summary is', summary)
    # 笔记本内 Markdown 渲染
    display(Markdown(summary))
````
