```python
import sys
sys.path.append('../')
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
from src.evaluator import compute_metrics
sns.set_style('whitegrid')
```

## 加载所有结果

```python
with open('../data/test_items_rag.pkl', 'rb') as f:
    test = pickle.load(f)
test_prices = np.array([item.price for item in test])
print(f"Loaded {len(test)} test items")
```

## 按价格范围划分的性能

```python
# 按价格区间分析表现
# Analyze performance by price bucket
buckets = [item.price_bucket for item in test]
bucket_analysis = pd.DataFrame({
    'bucket': buckets,
    'actual': test_prices
})
print(bucket_analysis.groupby('bucket')['actual'].describe())
```

## 按类别划分的表现

```python
# Analyze by category
categories = [item.category for item in test]
category_analysis = pd.DataFrame({
    'category': categories,
    'actual': test_prices
})
print(category_analysis.groupby('category')['actual'].describe())
```
