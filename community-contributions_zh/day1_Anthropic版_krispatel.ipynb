````python
# ========== 从 .env 加载环境变量并检查 ANTHROPIC_API_KEY ==========

# override=True：.env 覆盖进程里已有的同名变量
load_dotenv(override=True)
# 取出 Anthropic 密钥（变量名必须和 .env 里一致）
api_key = os.getenv('ANTHROPIC_API_KEY')

# ========== 检查 Key：常见配置错误提前打印提示（文案保留英文） ==========

# 没读到密钥
if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
# 前缀检查：原作者沿用了 sk-proj- 判断（Anthropic 密钥格式可能不同，以你实际密钥为准）
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
# 首尾空白字符
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")
````

# 先快速调用一次前沿模型作为预览！


````python
# ========== 预览：构造发给 Claude 的第一条 user 消息 ==========
# 有问题请看 Troubleshooting notebook。

# 自然语言问候（保留英文：这是发给模型的 prompt 内容）
message = "Hello, Claude! This is my first ever message to you! Hi!"

# Anthropic Messages API：user 消息放在 messages 列表里（system 通常单独传参）
messages = [{"role": "user", "content": message}]

# 在笔记本里直接显示该列表，确认结构正确
messages
````

````python
# ========== 创建 Anthropic 客户端并完成第一次非流式调用 ==========

# 实例化客户端：默认从环境变量读取 ANTHROPIC_API_KEY
claude = Anthropic()

# messages.create：同步调用 Claude；max_tokens 限制回复最长 token 数
response = claude.messages.create(model="claude-haiku-4-5", max_tokens=1024, messages=messages)
# content 是块列表；取第一块的 .text 即为助手回复正文
response.content[0].text
````

## 提示词类型

你可能已经知道——如果还不知道，很快就会非常熟悉！

像 GPT / Claude 这类模型，训练时就习惯按特定方式接收指令。

它们期望收到：

**System 提示**：告诉模型要执行什么任务、用什么语气

**User 提示**——对话开场，模型需要回复的内容


````python
# ========== 定义 system 提示（角色 + 语气 + 输出格式） ==========
# 稍后可实验，例如改成用西班牙语 Markdown 回复；正文保留英文以免改变模型行为

system_prompt = """
You are a snarky assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
````

````python
# ========== 定义 user 提示前缀：后面会拼上网页正文 ==========

user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""
````

## 消息（Messages）

OpenAI 的 API 期望按特定结构接收消息。
许多其他 API 也采用相同结构：

```python
[
    {"role": "system", "content": "system message goes here"},
    {"role": "user", "content": "user message goes here"}
]
```

Anthropic 略有不同：`system` 常常作为独立参数传入，`messages` 里主要放 user/assistant 轮次。

作为预览，接下来会做一次很简单的调用——我们还不会为难强大的模型（暂时！）


````python
# ========== 最小调用示例：system 单独传，messages 里只放 user ==========
# messages = [
#     {"role": "system", "content": "You are a helpful assistant"},
# ]

# system= 直接传系统指令；user 问一个极简算术题（字符串保留英文）
response = claude.messages.create(model="claude-haiku-4-5", max_tokens=1024, system="You are a snarky assistant", messages=[{"role": "user", "content": "What is 2 + 2?"}])
# 取出回复文本
response.content[0].text
````

## 现在用函数为 claude-haiku-4.5 构建有用的 messages


````python
# ========== 把 system + user 打包成 Anthropic 需要的 payload 字典 ==========

# website：已经抓取好的网页正文字符串
def messages_for(website):
    # 返回字典：system 用上面的 system_prompt；messages 里一条 user = 前缀 + 正文
    return {
         "system": system_prompt,
        "messages": [{"role": "user", "content": user_prompt_prefix + website}]
    }
````

````python
# ========== 试一下：用刚抓到的 ed 网页正文构造 payload ==========
# 可再换几个网站的抓取结果传入对比

messages_for(ed)
````

## 组合起来——Anthropic 的 API 非常简单！


````python
# ========== 端到端：URL → 抓取 → Claude 摘要 ==========
# 你很快会非常熟悉这类封装！

def summarize(url):
    # 1) 抓网页正文
    website = fetch_website_contents(url)
    # 2) 组装 system + messages
    payload = messages_for(website)
    # 3) 调用 Claude Messages API（非流式）
    response = claude.messages.create(
        model = "claude-haiku-4-5",
        messages = payload["messages"],
        max_tokens=1024
    )
    # 4) 返回助手文本
    return response.content[0].text
````

````python
# ========== 直接调用 summarize，在输出区看原始字符串 ==========
summarize("https://edwarddonner.com")
````

````python
# ========== 用 Markdown 漂亮展示输出 ==========

def display_summary(url):
    # 先拿到摘要字符串
    summary = summarize(url)
    # 再交给 IPython 渲染为排版后的 Markdown
    display(Markdown(summary))
````

````python
# ========== 一键展示：edwarddonner.com 的吐槽风摘要 ==========
display_summary("https://edwarddonner.com")
````
