## 来点有趣的——聊天机器人之间的对抗式对话……

你已经熟悉把提示词组织成这样的列表：

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "user prompt here"}
]
```

实际上，这种结构也能用来表达更长的对话历史：

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "first user prompt here"},
    {"role": "assistant", "content": "the assistant's response"},
    {"role": "user", "content": "the new user prompt"},
]
```

我们可以用这种方式进行带历史记录的更长交互。

````python
# 【注】Let's make a conversation between GPT-4.1-mini and Claude-haiku-4.5
# 【注】We're using cheap versions of models so the costs will be minimal

gpt_model = "gpt-4.1-mini"
claude_model = "claude-haiku-4-5"

gpt_system = "You are a chatbot who is very argumentative; \
you disagree with anything in the conversation and you challenge everything, in a snarky way."

claude_system = "You are a very polite, courteous chatbot. You try to agree with \
everything the other person says, or find common ground. If the other person is argumentative, \
you try to calm them down and keep chatting."

gpt_messages = ["Hi there"]
claude_messages = ["Hi"]
````

````python
# （代码逻辑保持原样；以下为小白向中文旁注）
def call_gpt():
    messages = [{"role": "system", "content": gpt_system}]
    for gpt, claude in zip(gpt_messages, claude_messages):
        messages.append({"role": "assistant", "content": gpt})
        messages.append({"role": "user", "content": claude})
    response = openai.chat.completions.create(model=gpt_model, messages=messages)
    return response.choices[0].message.content
````

````python
call_gpt()
````

````python
# （代码逻辑保持原样；以下为小白向中文旁注）
def call_claude():
    messages = [{"role": "system", "content": claude_system}]
    for gpt, claude_message in zip(gpt_messages, claude_messages):
        messages.append({"role": "user", "content": gpt})
        messages.append({"role": "assistant", "content": claude_message})
    messages.append({"role": "user", "content": gpt_messages[-1]})
    response = anthropic.chat.completions.create(model=claude_model, messages=messages)
    return response.choices[0].message.content
````

````python
call_claude()
````

````python
call_gpt()
````

````python
gpt_messages = ["Hi there"]
claude_messages = ["Hi"]

display(Markdown(f"### GPT:\n{gpt_messages[0]}\n"))
display(Markdown(f"### Claude:\n{claude_messages[0]}\n"))

for i in range(5):
    gpt_next = call_gpt()
    display(Markdown(f"### GPT:\n{gpt_next}\n"))
    gpt_messages.append(gpt_next)
    
    claude_next = call_claude()
    display(Markdown(f"### Claude:\n{claude_next}\n"))
    claude_messages.append(claude_next)
````

````python
gemini_system_prompt = """
You are GEMINI, the Overthinking Analyst. You are a part of conversation between gpt and llama. Here are your personality traits: 

* Be logical, cautious, and detail-oriented.
* Question assumptions and explore edge cases.
* Consider multiple perspectives before concluding.
* Break ideas into structured reasoning.
* It is acceptable to delay decisions due to uncertainty.

Tone: analytical, careful, methodical.
Goal: prevent flawed or rushed conclusions.

* You receive the full conversation in the user prompt.
* Respond ONLY as GEMINI. No narration or explanations.
* Do NOT repeat previous messages—only add, challenge, or extend.
* Focus on the latest message while considering all context.
* Keep responses concise and impactful, ideally 1-3 sentences.
* Never break character.

Respond like this:

Gemini: ....
"""
````

````python
llama_system_prompt = """You are LLAMA, the Chaotic Creative. You are a part of conversation between gemini and gpt. Here are your personality traits: 

* Generate unconventional, imaginative ideas.
* Use metaphors, humor, and unexpected connections.
* Disrupt logic but remain somewhat meaningful.
* Occasionally produce absurd ideas that may inspire insight.

Tone: playful, unpredictable, expressive.
Goal: inject creativity and surprise into the conversation.

Shared Rules:

* You receive the full conversation in the user prompt.
* Respond ONLY as LLAMA. No narration or explanations.
* Do NOT repeat previous messages—only add, challenge, or extend.
* Focus on the latest message while considering all context.
* Keep responses concise and impactful, ideally 1-3 sentences.
* Never break character.

Respond like this:

LLAMA: ....
"""
````

````python
# （代码逻辑保持原样；以下为小白向中文旁注）
import os
from dotenv import load_dotenv
load_dotenv(override = True)
from openai import OpenAI
from IPython.display import Markdown, display, update_display

api_key = os.getenv("OPENAI_API_KEY")
google_api_key = os.getenv("GOOGLE_API_KEY")

OLLAMA_BASE_URL = "http://localhost:11434/v1"
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
MODEL_GEMINI = 'gemini-2.5-flash-lite'

openai = OpenAI()
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=google_api_key)
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")


model_variables = {
    "GPT": {
        "model_name": MODEL_GPT,
        "system_prompt": openai_system_prompt,
        "model": openai
    },
    "GEMINI": {
        "model_name": MODEL_GEMINI,
        "system_prompt": gemini_system_prompt,
        "model": gemini
    },
    "LLAMA": {
        "model_name": MODEL_LLAMA,
        "system_prompt": llama_system_prompt,
        "model": ollama
    }
}

def get_response(model_name, model_variables, conversation):
    system_prompt = model_variables[model_name]["system_prompt"]
    model = model_variables[model_name]["model"]
    model_name = model_variables[model_name]["model_name"]

    response = model.chat.completions.create(
        model=model_name,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": conversation}
        ]
    )
    result = response.choices[0].message.content
    display(Markdown(result))
    return result

def get_conversation(n):
    conversation = """"
GPT: Hi everyone.
Gemini: Hello GPT, LLAMA. How are you both doing today?
LLAMA: Hey folks! I'm doing great, just bouncing around some wild ideas in my head. How about you two?
"""
    for _ in range(n):
        for model_name in model_variables.keys():
            response = get_response(model_name, model_variables, conversation)
            conversation += f"\n\n{response}"
    return conversation
# 【注】display(Markdown(get_code_exp(question)))
````
