### 把你的改动贡献到仓库，与他人分享

如果能收录你的贡献，我会非常感激。这能为其他学员增加价值，我也乐于看到！额外的好处是，你会在 Github 上作为仓库贡献者获得认可。

以下是发起 PR 的整体步骤与关键说明：  
https://edwarddonner.com/pr  

提交前请检查：  
1. 你的 PR 只包含 community-contributions 中的改动（除非我们已经讨论过）  
2. 所有 notebook 的输出都已清空  
3. 代码总计少于 2,000 行，且文件不要太多  
4. 不要包含不必要的测试文件，或过于冗长的 README、.env.example、emoji 或其他 LLM 产物！

非常感谢！
