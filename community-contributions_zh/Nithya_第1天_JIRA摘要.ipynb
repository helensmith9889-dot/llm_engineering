## 提示类型（Prompt Types）

你可能已经知道——如果还不知道，很快就会非常熟悉！

像 GPT 这样的模型经过训练，期望以特定方式接收指令。它们通常需要：

- **System prompt（系统提示）**：告诉模型当前任务、角色，以及应用什么语气/风格
- **User prompt（用户提示）**：对话的「开场」——模型需要回应的具体内容

后面我们会分别定义 `system_prompt` 与 `user_prompt_prefix`，再拼进 `messages`。


````python
# ========== 定义 system prompt：规定「怎么摘要、什么语气」 ==========

# 你可以稍后做实验，例如把最后一句改成要求用某种语言、某种风格回复
# system prompt 全文保留英文：改译会改变模型行为
system_prompt = """
You are a snarky assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
````

````python
# ========== 定义 user prompt 前缀：后面会拼上网页正文 ==========

# user 侧指令保留英文；真正的网站文本会在 messages_for() 里追加
user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""
````

## Messages（消息列表）

OpenAI 的 API 期望接收特定结构的 messages。许多其他厂商的 API 也共享类似结构：

```python
[
    {"role": "system", "content": "系统消息写在这里"},
    {"role": "user", "content": "用户消息写在这里"}
]
```

为给你一个预览，接下来会做一个相当简单的调用——我们还不会立刻上「最强」的模型。


````python
# ========== 迷你示例：system + user，问一个算术题 ==========

# 构造标准两段式 messages（英文 content 保留，影响模型回答）
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "What is 2 + 2?"}
]

# 用较小/较便宜的模型做一次完整调用，确认 messages 结构无误
response = openai.chat.completions.create(model="gpt-4.1-nano", messages=messages)
# 取出助手回复文本
response.choices[0].message.content
````

## 现在用函数为 GPT 构建「真正有用」的 messages

把上面的手工列表封装成 `messages_for(website)`：传入抓到的网页文本，自动带上我们定义的 system / user 前缀。


````python
# ========== messages_for：把网页文本塞进固定 prompt 模板 ==========

# 看看这个函数如何创建与上面完全相同的「列表套字典」格式
def messages_for(website):
    """返回 [system, user]；user 的 content = 前缀 + 网页正文。"""
    return [
        {"role": "system", "content": system_prompt},
        # 字符串拼接：前缀说明任务，website 是抓取结果
        {"role": "user", "content": user_prompt_prefix + website}
    ]
````

````python
# ========== 试一下 messages_for：用刚才抓到的 ed ==========

# 传入网页文本，查看拼好的 messages（可再换几个网站对比长度）
messages_for(ed)
````

## 把它们串起来——OpenAI 的 API 非常直接！

模式就是：`抓取 URL → messages_for → chat.completions.create → 取 content`。你会在本课里反复写这段。


````python
# ========== summarize：端到端「URL → 摘要字符串」 ==========

# 现在：调用 OpenAI API。你会对这个模式越来越熟！
def summarize(url):
    """抓取 url，调用模型，返回摘要文本（字符串）。"""
    # 1) 抓网页正文
    website = fetch_website_contents(url)
    # 2) 创建客户端请求：model id 保持可运行英文；messages 用上面的工厂函数
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(website)
    )
    # 3) 只返回助手消息正文，方便后续 display 或再处理
    return response.choices[0].message.content
````

````python
# ========== 试跑 summarize：对示例站点生成摘要字符串 ==========

# 直接调用；笔记本会显示返回的字符串
summarize("https://edwarddonner.com")
````

````python
# ========== display_summary：摘要 + Markdown 美化展示 ==========

# 把 summarize 的纯文本再用 Markdown 渲染，阅读体验更好
def display_summary(url):
    """对 url 做摘要，并在笔记本中以 Markdown 显示。"""
    # 先拿到摘要字符串
    summary = summarize(url)
    # display(Markdown(...))：在 Jupyter 里渲染标题/列表等
    display(Markdown(summary))
````

````python
# ========== 端到端展示：抓取 + 摘要 + Markdown ==========

# 对示例站点跑完整流程（需要有效 API Key 与网络）
display_summary("https://edwarddonner.com")
````
