## 提示词的类型

你可能已经知道了——如果还不知道，很快就会非常熟悉！

像 GPT 这样的模型被训练成以特定方式接收指令。

它们期望收到：

**系统提示词（system prompt）**，告诉它们正在执行什么任务、应使用什么语气

**用户提示词（user prompt）**——它们应回复的对话开场白


````python
# ========== 定义 system prompt：角色、语气、输出格式 ==========

# 可稍后改最后一句做实验，例如改成 Respond in markdown in Spanish.
# system_prompt 字符串本身保留英文（发给模型的指令）
system_prompt = """
You are a snarky assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
````

````python
# ========== 定义 user prompt 前缀：后面会拼上网页正文 ==========

# 前缀说明「下面是网站内容，请做短摘要」；真正正文在 messages_for 里拼接
user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""
````

## 消息（Messages）

OpenAI 的 API 期望以特定结构接收消息。
许多其他 API 也使用这种结构：

```python
[
    {"role": "system", "content": "system message goes here"},
    {"role": "user", "content": "user message goes here"}
]
```
先给你一个预览：接下来两个单元格会做一个相当简单的调用——我们还不会（暂时！）为难强大的 GPT


````python
# ========== 最小 messages 示例：system + user，再调一次 API ==========

# 演示结构：system 定助手身份，user 问「2+2」
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "What is 2 + 2?"}
]

# 用较小/较快的模型做结构验证
response = openai.chat.completions.create(model="gpt-4.1-nano", messages=messages)
# 打印助手回答
response.choices[0].message.content
````

## 现在让我们用函数为 GPT-4.1-mini 构建有用的消息


````python
# ========== messages_for：把「网页正文」包装成标准 messages ==========

# 返回值格式正是上面演示的 system + user 列表
def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        # user：前缀说明 + 抓取到的网站正文
        {"role": "user", "content": user_prompt_prefix + website}
    ]
````

````python
# ========== 试跑 messages_for：把刚才抓到的 ed 塞进去看看结构 ==========

# 可再换别的网站正文试几次
messages_for(ed)
````

## 是时候把它们整合起来了——OpenAI 的 API 非常简单！


````python
# ========== summarize：抓取 URL → 组 messages → 调 API → 返回摘要文本 ==========

# 你会反复写这种「抓取 + chat.completions.create」流水线
def summarize(url):
    # 1) 抓网页正文
    website = fetch_website_contents(url)
    # 2) 调用云端模型；messages_for 负责拼 system/user
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(website)
    )
    # 3) 返回助手生成的 Markdown 摘要字符串
    return response.choices[0].message.content
````

````python
# ========== 调用 summarize：对 edwarddonner.com 做一次摘要 ==========

summarize("https://edwarddonner.com")
````

````python
# ========== display_summary：summarize 后再用 Markdown 漂亮展示 ==========

def display_summary(url):
    # 先拿到摘要字符串
    summary = summarize(url)
    # 在笔记本输出区渲染 Markdown
    display(Markdown(summary))
````

````python
# ========== 端到端演示：抓取 + 摘要 + Markdown 显示 ==========

display_summary("https://edwarddonner.com")
````

# 我的作业：用 Ollama 在本地运行 LLM


````python
# ========== 作业导入：HTTP 探测 + OpenAI 兼容客户端 ==========

# requests：后面用 GET 检查本机 Ollama 是否在跑
import requests
# OpenAI SDK：也可指向 Ollama 的 OpenAI 兼容 /v1 端点
from openai import OpenAI


````

````python
# ========== 检查本机 Ollama 是否在监听 11434 ==========

# 若服务未启动会连接失败；成功则 .content 多为欢迎字节串
requests.get('http://localhost:11434').content
````

````python
# ========== 拉取要用的本地小模型（Jupyter ! 壳命令）==========

# ollama pull：从 Ollama 库下载 gemma3:270m 到本机
!ollama pull gemma3:270m
````

````python
# ========== 创建指向本地 Ollama 的 OpenAI 兼容客户端 ==========

# Ollama 的 OpenAI 兼容 Base URL（注意带 /v1）
ollama_base_url = 'http://localhost:11434/v1'

# api_key 对本地服务多为占位字符串；base_url 才决定打到哪
ollama = OpenAI(base_url = ollama_base_url, api_key = 'ollama')
# 看一眼客户端对象
ollama
````

````python
# ========== 为本地方案重写 messages_for（system/user 仍用英文 prompt）==========

def messages_for(website):
    '''message list for the website prompt'''

    # system：摘要助手角色与输出格式（发给模型，保留英文）
    system_prompt = """You are a helpful assistant that analyzes the contents of a website,
and provides a short, clear summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown."""

    # user 前缀：说明任务；后面拼接 website 正文
    user_prompt = """Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""

    return [
        {'role' : 'system', 'content' : system_prompt},
        {'role' : 'user', 'content' : user_prompt + website}
    ]

````

````python
# ========== summarize：抓取 URL，用本地 gemma3:270m 生成摘要 ==========

def summarize(url):
    '''
    Summarize the content of a website using Ollama.
    
    Args:
        url: The URL of the website to summarize
        
    Returns:
        A string containing the summary of the website
    '''

    # 复用本课抓取工具拿到正文
    content = fetch_website_contents(url)

    # 走本地 ollama 客户端（不是云端 openai 变量）；model 名需已 pull
    response = ollama.chat.completions.create(model = 'gemma3:270m', messages = messages_for(content))

    # 返回助手文本
    return response.choices[0].message.content
````

````python
# ========== 试跑：维基百科 Messi 页面摘要 ==========

summarize('https://en.wikipedia.org/wiki/Lionel_Messi')
````

````python
# ========== 包装展示：摘要结果用 Markdown 渲染 ==========

def background_summary(url):

    # 调用上面的 summarize
    summarized = summarize(url)

    # display(Markdown(...))：在笔记本里漂亮显示
    return display(Markdown(summarized))
````

````python
# ========== 端到端：本地 Ollama 摘要 + Markdown 显示 ==========

background_summary('https://en.wikipedia.org/wiki/Lionel_Messi')
````
