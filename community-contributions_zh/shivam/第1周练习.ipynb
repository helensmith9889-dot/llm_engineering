# 第 1 周末练习 —— Gemini 技术问答助手（Web3 / Blockchain）

## 练习目标（理念）

熟悉 **提示词（prompt）**、**消息角色** 以及 **Google Gemini API**，构建一个能回答技术问题、给出清晰简洁答案的小工具。

- **输入**：system 定角色（Web3 / Blockchain 助手）+ user 提具体问题
- **输出**：模型生成的技术解答（可 `print` 或用 Markdown 展示）
- **库**：`google.generativeai`

## 和本课 Day 1 / Day 2 的关系

| 本课概念 | 本练习里你会看到 |
|----------|------------------|
| System / User 提示 | `system_instruction` + `generate_content(user_prompt)` |
| 模型常量 | `MODEL = 'gemini-flash-latest'` |
| 环境变量 | `.env` 里的 `GOOGLE_API_KEY` |
| 列表可用模型 | `genai.list_models()` 过滤支持 `generateContent` 的条目 |

## 怎么跑

1. 准备 `.env`：写入 `GOOGLE_API_KEY`
2. 安装 `google-generativeai`、`python-dotenv` 等
3. 从上到下运行：配置 →（可选）列模型 → 调用 → 展示结果


```python
# ========== 导入：环境、展示、Gemini SDK ==========

# 导入标准库 os：从环境变量读取 GOOGLE_API_KEY
import os
# 从 dotenv 导入 load_dotenv：把 .env 里的密钥读进环境变量，避免把密钥写进代码
from dotenv import load_dotenv
# 从 IPython.display 导入 Markdown、display：在笔记本里漂亮渲染模型回答
from IPython.display import Markdown, display
# 导入 google.generativeai 并起别名 genai：Gemini 官方 Python SDK
import google.generativeai as genai

# 加载当前工作目录下的 .env（无 override 参数则不覆盖已有环境变量）
load_dotenv()
```

```python
# ========== 常量：模型名与 API Key 集中存放 ==========

# 要调用的 Gemini 模型 id；字符串必须与 Google AI 可用模型名一致
MODEL = 'gemini-flash-latest'
# 从环境变量读取 Google API Key（常见名 GOOGLE_API_KEY）
GOOGLE_API_KEY = os.getenv('GOOGLE_API_KEY')
```

```python
# ========== 配置 SDK：把 API Key 交给 genai 全局客户端 ==========

# configure：后续 list_models / GenerativeModel 都会用这个密钥鉴权
genai.configure(api_key=GOOGLE_API_KEY)
```

```python
# ========== 列出当前密钥可用、且支持 generateContent 的模型 ==========

# list_models：枚举账号可见的模型元数据
for m in genai.list_models():
    # 只打印支持文本生成（generateContent）的模型，过滤掉仅嵌入等类型
    if 'generateContent' in m.supported_generation_methods:
        # m.name 形如 models/gemini-2.5-flash；便于对照上面的 MODEL 常量
        print(m.name)
```

```python
# ========== 三个小函数：调 API / 打印 / Markdown 展示 ==========

# 封装一次生成：传入 system_prompt 与 user_prompt，返回纯文本
def make_api_call(system_prompt, user_prompt):
    # GenerativeModel：绑定模型名；system_instruction 相当于 Chat 里的 system 角色
    model = genai.GenerativeModel(model_name=MODEL, system_instruction=system_prompt)
    # generate_content：把 user_prompt 发给模型，阻塞直到整段生成完
    response = model.generate_content(user_prompt)
    # .text：取出助手回复的字符串
    return response.text

# 简单打印：适合终端对照或快速调试
def print_result(result):
    print(result)

# 用笔记本 Markdown 渲染：代码块、列表会更好看
def display_result(result):
    display(Markdown(result))
```

```python
# ========== 提问：Web3 助手 + ERC-20 示例请求 ==========

# system prompt 保留英文：这是发给模型的角色/规则，改译会改变回答风格
system_prompt = """
You are a Web 3 and Blockchain assistant.

Help with:
- Ethereum, Solidity, and smart contracts
- Decentralized applications (dApps)
- Cryptocurrencies and DeFi
- NFTs and token standards
- Blockchain security and best practices

Provide clear, practical, and production-ready answers.
Use examples when helpful.
Avoid unnecessary explanations.
Assume the user has technical experience.
"""

# user prompt 保留英文：真正的用户问题字符串，影响模型输出内容
user_prompt = """
Give me a Solidity smart contract for a simple ERC-20 token.
"""

# 调用封装函数：内部走 Gemini generate_content
result = make_api_call(system_prompt, user_prompt)

# 先用纯文本打印一份，方便在输出区直接复制
print_result(result)
```

```python
# ========== 同一份 result：再用 Markdown 漂亮展示一遍 ==========

# display_result：把上一格拿到的字符串渲染成 Markdown（代码围栏等会排版）
display_result(result)
```
